Kümülatif Beklenti Teorisi (Cumulative Prospect Theory, CPT)'ne göre Belirlilik Eşdeğeri (Certainty Equivalent) hesaplayan VBA (Visual Basic for Applications) kodudur. Şimdi parçalar halinde Türkçe olarak açıklayayım:

Kümülatif Beklenti Teorisi (CPT) Nedir?

CPT, insanların riskli kararlar alırken kazançları ve kayıpları farklı değerlendirdiğini ve olasılıkları çarpıtarak düşündüğünü varsayan bir davranışsal ekonomi teorisidir. Daniel Kahneman ve Amos Tversky tarafından geliştirilmiştir.



Kodun Temel Fonksiyonları

1. Değerleme Fonksiyonları (Value Functions)

Bu fonksiyonlar, kazançları ve kayıpları farklı şekilde değerlendirir

• Upos (a, alpha) Kazançlar için değer fonksiyonu
• Uneg (b, beta, lambda) Kayıplar için değer fonksiyonu

Parametreler

• α, β risk duyarlılığı (genelde 0 < α, β < 1)
• λ kayıplara verilen ağırlık (kayıp iğrençliği katsayısı)

2. Belirlilik Eşdeğeri (Certainty Equivalent)

Riskli bir seçeneğin birey için eşdeğer kesin değeri.

• CEpos (c, alpha) Kazançlar için belirlilik eşdeğeri
• CEneg (d, beta, lambda) Kayıplar için belirlilik eşdeğeri
• CertaintyEquivalent (x, alpha, beta, lambda) Genel belirlilik eş-değeri fonksiyonu



3. Ağırlık Fonksiyonları (Weighting Functions)

İnsanların olasılıkları çarpıtmasını modeller

• Wplus (z, gamma) Kazançlar için olasılık ağırlığı
• Wminus (t, delta) Kayıplar için olasılık ağırlığı

İstatistiksel Fonksiyonlar

Bu fonksiyonlar dağılımların özelliklerini hesaplar

• Mean(data(), p()) → Ağırlıklı ortalama
• SD(data(), p()) → Ağırlıklı standart sapma
• Skewness(data(), p()) → Basıklık
• Kurtosis(data(), p()) → Çarpıklık

Yardımcı Fonksiyonlar

• sumoutcomes(a, b, Outcomes()) Belirli bir aralıktaki sonuçların top-
lam olasılığını verir.

• Last(choice, rng) Excel’de son dolu hücreyi bulmak için kullanılır.

• Sub Calculate() (Eksik) Hesaplama sürecini başlatan alt prosedür.

Ne İşe Yarar?

Bu kod, örneğin bir yatırımın farklı sonuçları (kazanç ve kayıp senaryoları) için,
CPT’ye göre bir kişinin bu yatırımı nasıl değerlendireceğini ve yatırımın onun
gözünde ne kadar “kesin değer”e sahip olduğunu hesaplamak için kullanılabilir.

Örnek Senaryo
Bir piyango (lotto) seçeneği üzerinden CPT hesaplaması

Sonuç (TL),  Olasılık (%):
(+100, 40%)- (-50, 60%)


CPT Parametreleri

• α0.88 (kazanç için duyarlılık)

• β0.88 (kayıp için duyarlılık)

• λ2.25 (kayıp iğrençliği katsayısı)

• γ0.61 (kazanç olasılık ağırlığı)

• δ0.69 (kayıp olasılık ağırlığı)

Adım Adım Hesaplama

1. Değer Fonksiyonları


2. Olasılık Ağırlıkları


3. Toplam Beklenen Fayda


4. Belirlilik Eşdeğeri



```python
#  CPT

import numpy as np

# CPT parametreleri
alpha = 0.88
beta = 0.88
lambd = 2.25
gamma = 0.61
delta = 0.69

# Değer fonksiyonları
def U_pos(x, alpha=alpha):
    return x**alpha

def U_neg(x, beta=beta, lambd=lambd):
    return -lambd * ((-x)**beta)

# Olasılık ağırlık fonksiyonları
def W_plus(p, gamma=gamma):
    return p**gamma / (p**gamma + (1-p)**gamma)**(1/gamma)

def W_minus(p, delta=delta):
    return p**delta / (p**delta + (1-p)**delta)**(1/delta)

# Senaryo verileri
outcomes = [100, -50]
probabilities = [0.4, 0.6]

# Hesaplamalar
U_pos_val = U_pos(outcomes[0])
U_neg_val = U_neg(outcomes[1])
w_plus = W_plus(probabilities[0])
w_minus = W_minus(probabilities[1])

V = U_pos_val * w_plus + U_neg_val * w_minus

if V >= 0:
    CE = V**(1/alpha)
else:
    CE = -((-V)/lambd)**(1/beta)

print(f"Belirlilik Eşdeğeri: {CE:.2f} TL")
```
