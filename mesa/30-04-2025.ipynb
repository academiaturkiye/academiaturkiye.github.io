```python

# 1. Karar Teorileri

class DecisionTheory:
    def select_strategy(self, agent_state, model): # Added model parameter
        raise NotImplementedError

class ExpectedUtilityTheory(DecisionTheory):
    def select_strategy(self, agent_state, model): # Added model parameter
        if agent_state['risk_aversion'] > 0.5:
            return teknik1(model) # Pass model to strategy
        else:
            return temel(model) # Pass model to strategy

class ProspectTheory(DecisionTheory):
    def select_strategy(self, agent_state, model): # Added model parameter
        if agent_state['recent_loss'] > 0:
            return gurultu(model) # Pass model to strategy
        else:
            return zitlik(model) # Pass model to strategy
```

```python
# 3. Ajan (Mesa Agent)

import random

class DecisionAgent(mesa.Agent):
    def __init__(self, model, decision_theory):
        super().__init__(model)
        self.decision_theory = decision_theory
        #self.strategy = None
        self.state = {
            'risk_aversion': self.random.random(),
            'recent_loss': self.random.randint(0, 1)  # 0 veya 1
        }

          # Simülasyonun ilk fiyatı

        self.action = None

        self.last_action = None
        self.first_action = None  # İlk aksiyon: "buy" mı "sell" mi?
        self.pending_buy_price = self.model.price[-1]       # Alış fiyatını bekleme, başlangıç fiyatı olarak ayarla
        self.trade_count = 0


        self.price_history = []  # Alış-satış arasındaki fiyat değişimleri
        self.positive_returns = []
        self.negative_returns = []

        self.market_states = []  # Her adımda piyasa durumunu ("boğa", "ayı", "aralık") tutacak liste


        self.positive_return_avg = 0
        self.negative_return_avg = 0
        self.positive_return_freq = 0
        self.negative_return_freq = 0

        self.initial_price = self.model.price



        # Stratejiyi karar teorisinden seç
        self.strategy = self.decision_theory.select_strategy(self.state, model)
        print(f"Ajan {self.unique_id} oluşturuldu: Karar Teorisi: {decision_theory} Strateji: {self.strategy.__class__.__name__}")

    def step(self):
        #
        # Stratejiyi uygula
        self.agent_state = self.strategy.act(self.state)
        print(f"Ajan {self.unique_id} → Durum: {self.state} → Aksiyon: {self.agent_state}")
        print(f"Ajan {self.unique_id} → pending: {self.pending_buy_price}")

        # Eğer bir alış emri yapılmışsa (pending_buy_price None değilse),
        # her adımda sat fiyatına kadar fiyat değişikliğini kaydet


        if self.agent_state == "sell":
            self.pending_buy_price = None

        # pending_buy_price'ı sadece "buy" aksiyonunda güncelle:
        if self.agent_state == "buy":
            self.pending_buy_price = self.model.price[-1]  # Son fiyatı al


        if self.agent_state == "hold" and self.pending_buy_price is not None:
            # İki adım arasındaki fiyat farkını hesapla
            if len(self.model.price) > 1:  # En az iki fiyat verisi varsa
                price_change = self.model.price [-1] - self.model.price [-2]
            else:
                price_change = 0  # İlk adım için fiyat farkı 0 olarak ayarlanır

            self.price_history.append(price_change)
        print(f"Ajan {self.unique_id} → Fiyat Geçmişi (değişimler): {self.price_history}")
        return









        # Belirli bir olasılıkla karar teorisini değiştir
        if self.random.random() < 0.2:
            self.mutate_decision_theory()

    def mutate_decision_theory(self):
        old_theory = self.decision_theory.__class__.__name__
        if isinstance(self.decision_theory, ExpectedUtilityTheory):
            self.decision_theory = ProspectTheory()
        else:
            self.decision_theory = ExpectedUtilityTheory()
        print(f"Ajan {self.unique_id} karar teorisini değiştirdi: {old_theory} → {self.decision_theory.__class__.__name__}")

```

```python

# 4. Model (Mesa Model)

import random
class DecisionModel(mesa.Model):
    def __init__(self, num_agents):
        super().__init__()
        self.num_agents = num_agents
        self.mevcut_adim = 0

        self.price = [100]


        # Ajanları oluştur
        for i in range(self.num_agents):
            if self.random.random() < 0.5:
                theory = ExpectedUtilityTheory()
            else:
                theory = ProspectTheory()
            agent = DecisionAgent(self, theory)
            self.agents.add(agent)

    def step(self):
        self.mevcut_adim += 1
        self.price.append(self.price[-1] + random.randint(-5, 5))
        print(f"Şimdiki Fiyat: {self.price}")

        self.agents.do("step")
```

```python
class Agent:
    def __init__(self, unique_id, initial_price):
        self.unique_id = unique_id
        self.price_history = []  # Alış ve satışlar arasındaki fiyat değişimleri
        self.positive_returns = []
        self.negative_returns = []

        self.first_action = None  # İlk aksiyon: "buy" mı "sell" mi?
        self.pending_buy_price = None  # Alış fiyatı bekliyor mu?
        self.trade_count = 0

        self.initial_price = initial_price  # Simülasyonun ilk fiyatı

        self.positive_return_avg = 0
        self.negative_return_avg = 0
        self.positive_return_freq = 0
        self.negative_return_freq = 0

    def step(self, current_price, action):
        """
        Agent her adımda bir aksiyon alır: 'buy' ya da 'sell'.
        current_price: şimdiki piyasa fiyatı
        action: buy veya sell
        """

        # İlk aksiyon
        if self.first_action is None:
            self.first_action = action
            if action == "sell":
                self.pending_buy_price = self.initial_price  # İlk fiyatı alış gibi kabul et
            return

        # Aksiyon akışı
        if action == "buy":
            # Yeni alış yapıldı, bir şey yapmaya gerek yok (satış bekleyeceğiz)
            if self.pending_buy_price is None:
                self.pending_buy_price = current_price
        elif action == "sell" and self.pending_buy_price is not None:
            # Satış yapılıyor ve bir alış var
            price_change = current_price - self.pending_buy_price
            self.price_history.append(price_change)

            # Satıştan sonra pending_buy_price resetlenir (ta ki yeni alış gelene kadar)
            self.pending_buy_price = None
            self.trade_count += 1

    def analyze_returns(self):
        """
        Tüm kaydedilen price_change'leri analiz eder:
        Pozitif ve negatif getirilerin ortalaması ve frekansı hesaplanır.
        """
        for change in self.price_history:
            if change > 0:
                self.positive_returns.append(change)
            elif change < 0:
                self.negative_returns.append(change)

        if self.positive_returns:
            self.positive_return_avg = sum(self.positive_returns) / len(self.positive_returns)
            self.positive_return_freq = len(self.positive_returns)

        if self.negative_returns:
            self.negative_return_avg = sum(self.negative_returns) / len(self.negative_returns)
            self.negative_return_freq = len(self.negative_returns)

    def report(self):
        return {
            "Agent": self.unique_id,
            "Positive Return Average": self.positive_return_avg,
            "Positive Return Frequency": self.positive_return_freq,
            "Negative Return Average": self.negative_return_avg,
            "Negative Return Frequency": self.negative_return_freq
        }
```

```python
import random

class Agent:
    def __init__(self, unique_id, initial_price, current_step):
        self.unique_id = unique_id
        self.price_history = []  # Alış-satış arasındaki fiyat değişimleri
        self.positive_returns = []
        self.negative_returns = []

        self.first_action = None
        self.pending_buy_price = None
        self.trade_count = 0

        self.initial_price = initial_price
        self.positive_return_avg = 0
        self.negative_return_avg = 0
        self.positive_return_freq = 0
        self.negative_return_freq = 0

        # Emir zamanı ayarları
        self.next_action_step = current_step + random.randint(5, 15)  # İlk emir vereceği adım
        self.last_action = None  # Önceki aksiyonu kaydetmek için

        self.market_states = []  # Her adımda piyasa durumunu ("boğa", "ayı", "aralık") tutacak liste

    def step(self, current_step, current_price, market_state):
        """
        Ajan sadece belirli adımlarda aksiyon alır.
        Alış ve satış emirleri arasında geçen her adımda fiyat değişiklikleri price_change değişkenine atanır.
        market_state parametresi, o adımda piyasanın durumunu ("boğa", "ayı", "aralık") belirtir.
        """
        self.market_states.append(market_state)  # Piyasa durumunu listeye ekle

        if current_step < self.next_action_step:
            return  # Şu an bekleme zamanı, hiçbir şey yapmıyoruz

        # Şu an aksiyon zamanı geldi
        action = self.decide_action()

        if self.first_action is None:
            self.first_action = action
            if action == "sell":
                self.pending_buy_price = self.initial_price
            self.set_next_action_step(current_step)
            self.last_action = action
            return

        if action == "buy":
            if self.pending_buy_price is None:
                self.pending_buy_price = current_price
        elif action == "sell" and self.pending_buy_price is not None:
            price_change = current_price - self.pending_buy_price
            self.price_history.append(price_change)
            self.pending_buy_price = None
            self.trade_count += 1

        self.set_next_action_step(current_step)
        self.last_action = action

    def decide_action(self):
        """
        Basit rastgele bir 'buy' veya 'sell' kararı verir.
        Dilersen buraya karar teorilerini veya stratejiyi entegre edebiliriz.
        """
        if self.last_action == "buy":
            return "sell"
        else:
            return "buy"

    def set_next_action_step(self, current_step):
        """
        Bir sonraki emir adımını belirle (örneğin 5-15 adım sonrasına).
        """
        self.next_action_step = current_step + random.randint(5, 15)

    def analyze_returns(self):
        """
        Kaydedilen fiyat değişimlerini analiz eder ve piyasa durumlarına göre getirileri hesaplar.
        """
        self.positive_returns_by_state = {"boğa": [], "ayı": [], "aralık": []}
        self.negative_returns_by_state = {"boğa": [], "ayı": [], "aralık": []}

        for i, change in enumerate(self.price_history):
            state = self.market_states[i]  # O adımda piyasanın durumunu al
            if change > 0:
                self.positive_returns_by_state[state].append(change)
            elif change < 0:
                self.negative_returns_by_state[state].append(change)

        # Piyasa durumlarına göre getiri ortalamalarını ve frekanslarını hesapla
        self.positive_return_avg_by_state = {}
        self.positive_return_freq_by_state = {}
        self.negative_return_avg_by_state = {}
        self.negative_return_freq_by_state = {}

        for state in ["boğa", "ayı", "aralık"]:
            if self.positive_returns_by_state[state]:
                self.positive_return_avg_by_state[state] = sum(self.positive_returns_by_state[state]) / len(self.positive_returns_by_state[state])
                self.positive_return_freq_by_state[state] = len(self.positive_returns_by_state[state])
            else:
                self.positive_return_avg_by_state[state] = 0
                self.positive_return_freq_by_state[state] = 0

            if self.negative_returns_by_state[state]:
                self.negative_return_avg_by_state[state] = sum(self.negative_returns_by_state[state]) / len(self.negative_returns_by_state[state])
                self.negative_return_freq_by_state[state] = len(self.negative_returns_by_state[state])
            else:
                self.negative_return_avg_by_state[state] = 0
                self.negative_return_freq_by_state[state] = 0

    def report(self):
        """
        Ajanın performansını özetleyen bir rapor oluşturur.
        Piyasa durumlarına göre getirileri içerir.
        """
        report = {
            "Agent": self.unique_id,
            # ... (Diğer rapor bilgileri)
        }

        for state in ["boğa", "ayı", "aralık"]:
            report[f"{state.capitalize()} Positive Return Average"] = self.positive_return_avg_by_state[state]
            report[f"{state.capitalize()} Positive Return Frequency"] = self.positive_return_freq_by_state[state]
            report[f"{state.capitalize()} Negative Return Average"] = self.negative_return_avg_by_state[state]
            report[f"{state.capitalize()} Negative Return Frequency"] = self.negative_return_freq_by_state[state]

        return report
```
