```python
# Has multi-dimensional arrays and matrices.
# Has a large collection of mathematical functions to operate on these arrays.
import numpy as np

# Data manipulation and analysis.
import pandas as pd

# Data visualization tools.
import seaborn as sns

import mesa
```

```python
import random
from mesa.datacollection import DataCollector
import matplotlib.pyplot as plt



class BaseAgent(mesa.Agent):
    """Tüm ajanların ortak özellikleri"""
    def __init__(self,  model):
        super().__init__(model)


    def log(self, message):
        """Ortak loglama metodu"""
        print(f"[{self.__class__.__name__} {self.unique_id}] {message}")

class Employee(BaseAgent):
    """Çalışan ajanı"""
    def __init__(self, model, department):
        super().__init__(model)
        self.department = department
        self.task_completed = 0

    def step(self):
        if random.random() < 0.7:  # %70 görev tamamlama ihtimali
            self.task_completed += 1
            self.log(f"{self.department} departmanında görev tamamladı (Toplam: {self.task_completed})")
        else:
            self.log("Mola veriyor...")

class Manager(BaseAgent):
    """Yönetici ajanı"""
    def __init__(self, model, authority_level):
        super().__init__(model)
        self.authority_level = authority_level  # 1-3 arası

    def step(self):
        action = random.choice(["toplantı yapıyor", "rapor inceliyor", "strateji belirliyor"])
        self.log(f"{self.authority_level}. seviye yönetici {action}")

class Customer(BaseAgent):
    """Müşteri ajanı"""
    def __init__(self, model, customer_type):
        super().__init__(model)
        self.type = customer_type  # "bireysel" veya "kurumsal"
        self.satisfaction = 100

    def step(self):
        self.satisfaction += random.randint(-10, 15)
        self.satisfaction = max(0, min(100, self.satisfaction))
        self.log(f"{self.type} müşteri (Memnuniyet: {self.satisfaction}%)")

class CompanyModel(mesa.Model):
    def __init__(self, n, N_employees=0.4, N_managers=0.4, N_customers=0.2, seed=None):
        super().__init__(seed=seed)
        self.num_agents = n
        self.current_id = 0



        # Ajan sayılarını belirleme
        num_emp = int(self.num_agents * N_employees)
        num_man = int(self.num_agents * N_managers)
        num_cus = int(self.num_agents - (num_emp + num_man))  # Geriye kalanlar noise olur

        # Departmanlar
        departments = ["Üretim", "Pazarlama", "AR-GE"]
        # Çalışanlar
        for _ in range(num_emp):
            employee = Employee(self, department=random.choice(departments))
            self.agents.add(employee)

        # Yöneticiler
        for _ in range(num_man):
            manager = Manager(self, authority_level=random.randint(1, 3))
            self.agents.add(manager)

        # Müşteriler
        for _ in range(num_cus):
            customer = Customer(self, customer_type=random.choice(["bireysel", "kurumsal"]))
            self.agents.add(customer)

    def next_id(self):
        self.current_id += 1
        return self.current_id - 1

    def step(self):
        print(f"\n=== Adım  ===")
        self.agents.do("step")
```

```python

# Simülasyonu başlat
model = CompanyModel(n=10, N_employees=0.4, N_managers=0.4, N_customers=0.2)
for _ in range(3):
    model.step()
```
