-----
**bir finansal piyasada sadece gürültü tacirleri varsa, bu tacirler nasıl al-sat kararı verir ve fiyat oluşumunda etkileri nedir**

Gürültü tüccarları (noise traders), finansal piyasalarda temel analiz veya rasyonel bilgiler yerine duygusal, psikolojik veya rastgele faktörlerle işlem yapan katılımcılardır. Piyasada sadece gürültü tüccarları varsa, alım-satım kararları ve fiyat oluşumu şu şekilde etkilenir:

1. Gürültü Tüccarlarının Alım-Satım Kararları
Gürültü tüccarları, rasyonel temellere dayanmayan davranışlarla hareket eder:

Gösterge ve Yanılsamalara Dayalı Kararlar: Geçmiş fiyat hareketlerini yanlış yorumlayarak (örneğin, "aşırı alım/aşırı satım" gibi teknik göstergeleri abartılı şekilde kullanma) veya medya haberlerine aşırı tepki vererek işlem yaparlar.

Sürü Psikolojisi (Herd Behavior): Diğer gürültü tüccarlarının hareketlerini takip ederler, bu da fiyat balonları veya çöküşlere yol açabilir.

Aşırı Güven (Overconfidence): Rastgele kazançları beceri olarak yorumlayıp riski göz ardı edebilirler.

Rastgele Alım-Satım: Bazıları tamamen rastgele (örneğin, zar atarak) işlem yapabilir.

2. Fiyat Oluşumuna Etkileri
Gürültü tüccarlarının hakim olduğu bir piyasada:

Fiyatlar Temel Değerlerden Sapar: Piyasada bilgiyi doğru işleyen rasyonel yatırımcılar olmadığı için fiyatlar, varlıkların gerçek değeri (fundamental değer) ile ilişkisiz hale gelir. Örneğin, bir hisse senedinin fiyatı, şirketin kârından bağımsız olarak aşırı yükselebilir veya düşebilir.

Aşırı Oynaklık (Volatilite): Duygusal tepkiler ve rastgele işlemler, fiyatlarda nedensiz ve şiddetli dalgalanmalara yol açar.

Kalıcı Fiyat Çarpıklıkları: Gürültü tüccarlarının sistematik hataları (örneğin, tümünün aynı anda aşırı iyimser veya kötümser olması), fiyatların uzun süre gerçek değerinden uzak kalmasına neden olur.

Piyasa Verimsizliği: Bilginin fiyatlara yansımadığı bir piyasa oluşur, bu da kaynak dağılımını bozar.

3. Örnek Senaryo
Balon ve Çöküş: Gürültü tüccarları bir varlığa (örneğin, bir kripto para) topluca yönelirse, fiyat suni olarak yükselir (balon). Daha sonra aynı grup nedensiz bir panikle satışa geçerse fiyat çökebilir.

Rastgele Dalgalanma: Alım-satım kararları rastgele olduğunda, fiyatlar bir "random walk" (rastgele yürüyüş) sergiler, trend veya anlamlı bir model olmaz.

4. Teorik Çerçeve
Noise Trader Theory (Gürültü Tüccarı Teorisi): Fischer Black (1986), gürültü tüccarlarının piyasalarda likidite sağladığını ancak fiyatları bozduğunu savunur.

Davranışsal Finans: Gürültü tüccarlarının psikolojik önyargıları (örneğin, aşırı tepki, temsil etkisi), fiyat anomalilerini açıklar.

Sonuç
Sadece gürültü tüccarlarından oluşan bir piyasa, aşırı spekülatif, verimsiz ve öngörülemez olur. Fiyatlar gerçek değerle ilişkisini kaybeder ve piyasa, rasyonel yatırımcıların olmaması nedeniyle düzeltme mekanizmalarından yoksun kalır. Bu tür piyasalar, genellikle manipülasyonlara ve ani krizlere açıktır.

```python
# Python'da dört gürültü tüccarı ajanını modelleyen bir sınıf yapısı ve simülasyon kodu:


import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from enum import Enum

class Action(Enum):
    HOLD = 0
    BUY = 1
    SELL = -1

class NoiseTraderType(Enum):
    INDICATOR_BASED = 1
    HERD_BEHAVIOR = 2
    OVERCONFIDENT = 3
    RANDOM = 4

class NoiseTrader:
    def __init__(self, trader_type):
        self.type = trader_type
        self.profit_history = []
        self.position = 0
        self.last_action = Action.HOLD

    def decide(self, market_data, other_traders=None):
        if self.type == NoiseTraderType.INDICATOR_BASED:
            return self._indicator_based_decision(market_data)
        elif self.type == NoiseTraderType.HERD_BEHAVIOR:
            return self._herd_behavior_decision(other_traders)
        elif self.type == NoiseTraderType.OVERCONFIDENT:
            return self._overconfident_decision()
        elif self.type == NoiseTraderType.RANDOM:
            return self._random_decision()
        return Action.HOLD

    def _indicator_based_decision(self, market_data):
        # Basit RSI ve haber tepkisi simülasyonu
        current_rsi = market_data.get('rsi', 50)
        news_sentiment = market_data.get('news', 0)  # -1 (neg), 0 (nötr), 1 (poz)

        if current_rsi > 70 or news_sentiment < 0:
            return Action.SELL
        elif current_rsi < 30 or news_sentiment > 0:
            return Action.BUY
        return Action.HOLD

    def _herd_behavior_decision(self, other_traders):
        if not other_traders:
            return Action.HOLD

        # Son 5 işleme bak
        recent_actions = [t.last_action for t in other_traders[-5:]]
        buy_count = sum(1 for a in recent_actions if a == Action.BUY)
        sell_count = sum(1 for a in recent_actions if a == Action.SELL)

        if buy_count / len(recent_actions) > 0.7:
            return Action.BUY
        elif sell_count / len(recent_actions) > 0.7:
            return Action.SELL
        return Action.HOLD

    def _overconfident_decision(self):
        # Son 3 işlemin kâr/zarar durumu
        if len(self.profit_history) < 3:
            return Action.HOLD

        last_three = self.profit_history[-3:]
        profit = sum(last_three)

        if profit > 0:
            return Action.BUY  # 2x yerine basitleştirilmiş
        elif profit < 0:
            return Action.SELL  # Hatalı ama kendine güvenen karar
        return Action.HOLD

    def _random_decision(self):
        return np.random.choice([Action.BUY, Action.SELL, Action.HOLD], p=[0.4, 0.4, 0.2])

    def update_profit(self, profit):
        self.profit_history.append(profit)

class MarketSimulator:
    def __init__(self):
        self.price_history = [100]  # Başlangıç fiyatı
        self.traders = self._initialize_traders()

    def _initialize_traders(self):
        return [
            NoiseTrader(NoiseTraderType.INDICATOR_BASED),
            NoiseTrader(NoiseTraderType.HERD_BEHAVIOR),
            NoiseTrader(NoiseTraderType.OVERCONFIDENT),
            NoiseTrader(NoiseTraderType.RANDOM),
            NoiseTrader(NoiseTraderType.RANDOM)  # Ek rastgele tüccar
        ]

    def generate_market_data(self):
        # RSI 30-70 arasında rastgele hareket etsin
        rsi = np.random.uniform(20, 80)
        # Haber etkisi (%20 olasılıkla)
        news = np.random.choice([-1, 0, 1], p=[0.1, 0.8, 0.1])
        return {'rsi': rsi, 'news': news}

    def run_simulation(self, steps=100):
        for _ in range(steps):
            market_data = self.generate_market_data()
            current_price = self.price_history[-1]
            actions = []

            for trader in self.traders:
                action = trader.decide(market_data, self.traders)
                actions.append(action)
                trader.last_action = action

            # Fiyatı belirle (alım/satım dengesine göre)
            net_demand = sum(a.value for a in actions)
            new_price = current_price * (1 + net_demand * 0.01)  # %1 etki
            self.price_history.append(new_price)

            # Tüccarların kârını güncelle (basitleştirilmiş)
            price_change = new_price - current_price
            for i, trader in enumerate(self.traders):
                if actions[i] == Action.BUY:
                    trader.update_profit(price_change)
                elif actions[i] == Action.SELL:
                    trader.update_profit(-price_change)

    def plot_results(self):
        plt.figure(figsize=(12, 6))
        plt.plot(self.price_history, label='Fiyat')
        plt.title('Gürültü Tüccarları ile Piyasa Simülasyonu')
        plt.xlabel('Zaman Adımı')
        plt.ylabel('Fiyat')
        plt.legend()
        plt.grid(True)
        plt.show()

# Simülasyonu çalıştır
simulator = MarketSimulator()
simulator.run_simulation(steps=200)
simulator.plot_results()
```

```python
import numpy as np
import matplotlib.pyplot as plt

from mesa.space import MultiGrid
from mesa.datacollection import DataCollector

class NoiseTrader(mesa.Agent):
    """Gürültü tüccarı ajanı"""
    def __init__(self, model, trader_type):
        super().__init__(model)
        self.type = trader_type  # 1: Gösterge temelli, 2: Sürü, 3: Aşırı güven, 4: Rastgele
        self.profit_history = []
        self.last_action = 0  # 0: Hold, 1: Buy, -1: Sell
        self.position = 0

    def step(self):
        """Her adımda ajanın karar verme mekanizması"""
        market_data = self.model.get_market_data()
        other_traders = self.model.agents

        if self.type == 1:  # Gösterge temelli
            current_rsi = market_data['rsi']
            news = market_data['news']

            if current_rsi > 70 or news < 0:
                self.last_action = -1  # Sell
            elif current_rsi < 30 or news > 0:
                self.last_action = 1   # Buy
            else:
                self.last_action = 0    # Hold

        elif self.type == 2:  # Sürü davranışı
            recent_actions = [a.last_action for a in other_traders[-5:]]
            buy_count = sum(1 for a in recent_actions if a == 1)
            sell_count = sum(1 for a in recent_actions if a == -1)

            if buy_count / len(recent_actions) > 0.7:
                self.last_action = 1
            elif sell_count / len(recent_actions) > 0.7:
                self.last_action = -1
            else:
                self.last_action = 0

        elif self.type == 3:  # Aşırı güven
            if len(self.profit_history) >= 3:
                last_three = self.profit_history[-3:]
                if sum(last_three) > 0:
                    self.last_action = 1
                else:
                    self.last_action = -1
            else:
                self.last_action = 0

        else:  # Rastgele
            self.last_action = np.random.choice([-1, 0, 1], p=[0.4, 0.2, 0.4])

        # Pozisyonu güncelle
        if self.last_action == 1:
            self.position += 1
        elif self.last_action == -1:
            self.position -= 1

class MarketModel(mesa.Model):
    """Piyasa simülasyon modeli"""
    def __init__(self, N=20, width=10, height=10):
        super().__init__()

        self.num_agents = N
        self.grid = MultiGrid(width, height, True)

        self.price_history = [100]
        self.current_step = 0

        # Ajanları oluştur
        trader_types = [1, 2, 3, 4] * (N // 4)  # Eşit dağılım
        np.random.shuffle(trader_types)

        for i in range(self.num_agents):
            trader = NoiseTrader(self, trader_types[i])
            self.agents.add(trader)

            # Rastgele bir hücreye yerleştir
            x = self.random.randrange(self.grid.width)
            y = self.random.randrange(self.grid.height)
            self.grid.place_agent(trader, (x, y))

        # Veri toplayıcı
        self.datacollector = DataCollector(
            agent_reporters={"Position": "position"},
            model_reporters={"Price": "current_price"}
        )

    def get_market_data(self):
        """Rastgele piyasa verisi oluştur"""
        return {
            'rsi': np.random.uniform(20, 80),
            'news': np.random.choice([-1, 0, 1], p=[0.1, 0.8, 0.1])
        }

    @property
    def current_price(self):
        return self.price_history[-1]

    def step(self):
        """Modelin bir adımını çalıştır"""
        # Ajanların karar vermesi
        self.agents.do("step")

        # Toplam net talep
        actions = [a.last_action for a in self.agents]
        net_demand = sum(actions)

        # Fiyatı güncelle (%1 etki)
        new_price = self.current_price * (1 + net_demand * 0.01)
        self.price_history.append(new_price)

        # Kârları güncelle
        price_change = new_price - self.price_history[-2]
        for trader in self.agents:
            if trader.last_action == 1:  # Buy
                trader.profit_history.append(price_change)
            elif trader.last_action == -1:  # Sell
                trader.profit_history.append(-price_change)

        # Veri topla
        self.datacollector.collect(self)
        self.current_step += 1

def run_simulation(steps=200):
    """Simülasyonu çalıştır ve sonuçları göster"""
    model = MarketModel(N=20)
    for _ in range(steps):
        model.step()

    # Sonuçları görselleştir
    plt.figure(figsize=(12, 6))
    plt.plot(model.price_history)
    plt.title("Gürültü Tüccarları ile Piyasa Dinamikleri")
    plt.xlabel("Zaman Adımı")
    plt.ylabel("Fiyat")
    plt.grid(True)
    plt.show()

    return model

if __name__ == "__main__":
    model = run_simulation()
```

Bu satırlar, NoiseTrader (Gürültü Tüccarı) ajanının step() metodunun içinde yer alır ve ajanın aldığı karara (self.last_action) göre elindeki varlık miktarını (self.position) günceller.

Pozisyonu güncelle: Bu bir yorum satırıdır ve bu bloğun amacını belirtir: Ajana ait position (pozisyon/elindeki miktar) bilgisini güncellemek.

if self.last_action == 1:: Bu satır, ajanın bir önceki adımda aldığı kararın 1 olup olmadığını kontrol eder. Daha önceki kodda self.last_action = 1 değeri "Alım" (Buy) aksiyonunu temsil ediyordu. Yani bu satır "Eğer ajan Alım kararı verdiyse..." anlamına gelir.

self.position += 1: Eğer yukarıdaki if koşulu doğruysa (ajan Alım kararı verdiyse), bu satır ajanın self.position özniteliğinin değerini 1 artırır. Bu basit modelde position, ajanın elinde tuttuğu varlık miktarını temsil eder. Alım yapmak, ajanın elindeki varlık miktarını artırır.

elif self.last_action == -1:: Bu satır, ajanın bir önceki adımda aldığı kararın -1 olup olmadığını kontrol eder. self.last_action = -1 değeri "Satım" (Sell) aksiyonunu temsil ediyordu. Bu satır, ilk if koşulu yanlışsa ("Eğer ajan Alım yapmadıysa") kontrol edilir ve "Eğer ajan Satım kararı verdiyse..." anlamına gelir.

self.position -= 1: Eğer yukarıdaki elif koşulu doğruysa (ajan Satım kararı verdiyse), bu satır ajanın self.position özniteliğinin değerini 1 azaltır. Satım yapmak, ajanın elindeki varlık miktarını azaltır.

Eğer self.last_action değeri 0 (Bekle/Hold) ise, ne if ne de elif koşulu doğru olmaz ve self.position değeri bu adımda değişmez.

Ajanlar portföyündeki nakit ve pozisyonlarını kontrol ediyorlar ve al-sat sonrası portföylerinde güncelleme yapıyorlar.




```python

import matplotlib.pyplot as plt

from mesa.space import MultiGrid
from mesa.datacollection import DataCollector

class NoiseTrader(mesa.Agent):
    """Geliştirilmiş gürültü tüccarı ajanı"""
    def __init__(self,model, trader_type):
        super().__init__(model)
        self.type = trader_type  # 1: Gösterge, 2: Sürü, 3: Aşırı güven, 4: Rastgele
        self.cash = 10000  # Başlangıç nakiti
        self.positions = 10  # Başlangıç pozisyon miktarı
        self.confidence = np.random.uniform(0.5, 1.0)  # Aşırı güven seviyesi (0.5-1.0)
        self.herd_tendency = np.random.uniform(0.3, 0.8) if trader_type == 2 else 0  # Sürü eğilimi
        self.trade_history = []

    def determine_trade_amount(self, price):
        """Güven seviyesine göre alım/satım miktarını belirle"""
        base_amount = 1  # Minimum işlem miktarı
        max_possible = min(self.cash / price if self.cash > 0 else 0,
                          self.positions if self.positions > 0 else 0)

        if self.type == 3:  # Aşırı güvenenler
            return min(base_amount + int(5 * self.confidence), max_possible)
        elif self.type == 2:  # Sürü davranışı
            return min(base_amount + int(3 * self.herd_tendency), max_possible)
        else:
            return base_amount

    def execute_trade(self, action, price):
        """Alım/satım işlemini gerçekleştir"""
        trade_amount = self.determine_trade_amount(price)

        if action == 1 and self.cash >= price * trade_amount:  # Alım
            self.positions += trade_amount
            self.cash -= price * trade_amount
            self.trade_history.append(('BUY', trade_amount, price))
            return trade_amount

        elif action == -1 and self.positions >= trade_amount:  # Satım
            self.positions -= trade_amount
            self.cash += price * trade_amount
            self.trade_history.append(('SELL', trade_amount, price))
            return -trade_amount

        return 0

    def step(self):
        """Güncellenmiş karar mekanizması"""
        price = self.model.current_price
        market_data = self.model.get_market_data()
        other_traders = self.model.agents

        action = 0  # 0: Hold, 1: Buy, -1: Sell

        if self.type == 1:  # Gösterge temelli
            if market_data['rsi'] > 70 or market_data['news'] < 0:
                action = -1
            elif market_data['rsi'] < 30 or market_data['news'] > 0:
                action = 1

        elif self.type == 2:  # Sürü davranışı
            recent_actions = [a.last_action for a in other_traders[-5:]]
            buy_ratio = sum(1 for a in recent_actions if a == 1) / len(recent_actions)

            if buy_ratio > (0.5 + self.herd_tendency/2):
                action = 1
            elif buy_ratio < (0.5 - self.herd_tendency/2):
                action = -1

        elif self.type == 3:  # Aşırı güven
            if len(self.trade_history) > 0:
                last_trade = self.trade_history[-1]
                if last_trade[0] == 'BUY' and price > last_trade[2] * 1.02:
                    action = 1  # Kazanıyorsa daha fazla al
                elif last_trade[0] == 'SELL' and price < last_trade[2] * 0.98:
                    action = -1  # Kazanıyorsa daha fazla sat
                else:
                    action = 1 if np.random.random() < self.confidence else -1

        else:  # Rastgele
            action = np.random.choice([-1, 0, 1], p=[0.4, 0.2, 0.4])

        # İşlemi gerçekleştir
        trade_impact = self.execute_trade(action, price)
        self.last_action = action

        # Model için işlem etkisini kaydet
        self.model.trade_impacts.append(trade_impact * price)

class MarketModel(mesa.Model):
    """Geliştirilmiş piyasa modeli"""
    def __init__(self, N=20, width=10, height=10):
        super().__init__()
        self.num_agents = N
        self.grid = MultiGrid(width, height, True)

        self.price_history = [100]
        self.trade_impacts = []
        self.current_step = 0

        # Ajanları oluştur (eşit dağılım)
        trader_types = [1, 2, 3, 4] * (N // 4)
        np.random.shuffle(trader_types)

        for i in range(self.num_agents):
            trader = NoiseTrader(self, trader_types[i])
            self.agents.add(trader)
            x, y = self.random.randrange(width), self.random.randrange(height)
            self.grid.place_agent(trader, (x, y))

        # Gelişmiş veri toplama
        self.datacollector = DataCollector(
            agent_reporters={
                "Cash": "cash",
                "Positions": "positions",
                "Type": "type"
            },
            model_reporters={
                "Price": "current_price",
                "Volume": "current_volume"
            }
        )

    def get_market_data(self):
        """Piyasa verilerini oluştur"""
        return {
            'rsi': np.random.uniform(20, 80),
            'news': np.random.choice([-1, 0, 1], p=[0.1, 0.8, 0.1])
        }

    @property
    def current_price(self):
        return self.price_history[-1]

    @property
    def current_volume(self):
        return len(self.trade_impacts)  # Basit hacim ölçümü

    def step(self):
        """Model adımı"""
        self.trade_impacts = []  # Yeni adım için temizle

        # Ajanların işlem yapması
        self.agents.do("step")

        # Fiyatı güncelle (arz-talep dengesine göre)
        net_impact = sum(self.trade_impacts)
        price_change = net_impact * 0.0001  # Piyasa derinliği ayarı
        new_price = self.current_price * (1 + price_change)

        # Fiyat sınırlamaları
        new_price = max(1, min(1000, new_price))  # 1-1000 aralığı
        self.price_history.append(new_price)

        # Veri topla
        self.datacollector.collect(self)
        self.current_step += 1

def run_simulation(steps=200):
    """Simülasyonu çalıştır ve sonuçları göster"""
    model = MarketModel(N=20)

    for _ in range(steps):
        model.step()

    # Sonuçları görselleştir
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 10))

    # Fiyat grafiği
    ax1.plot(model.price_history)
    ax1.set_title("Piyasa Fiyat Hareketleri")
    ax1.set_xlabel("Zaman Adımı")
    ax1.set_ylabel("Fiyat")
    ax1.grid(True)

    # Ajan varlıkları
    agent_data = model.datacollector.get_agent_vars_dataframe()
    last_step = agent_data.index.get_level_values('Step').max()
    last_data = agent_data.xs(last_step, level='Step')

    for trader_type in [1, 2, 3, 4]:
        type_data = last_data[last_data['Type'] == trader_type]
        ax2.scatter(type_data['Cash'], type_data['Positions'],
                   label=f"Tip {trader_type}", alpha=0.6)

    ax2.set_title("Ajan Portföyleri (Son Durum)")
    ax2.set_xlabel("Nakit")
    ax2.set_ylabel("Pozisyon Miktarı")
    ax2.legend()
    ax2.grid(True)

    plt.tight_layout()
    plt.show()

    return model

if __name__ == "__main__":
    model = run_simulation(steps=200)
```

```python
#yukarıdaki hata veren kodun düzeltilmiş hali

import matplotlib.pyplot as plt

from mesa.space import MultiGrid
from mesa.datacollection import DataCollector

class NoiseTrader(mesa.Agent):
    """Geliştirilmiş gürültü tüccarı ajanı"""
    def __init__(self,model, trader_type):
        super().__init__(model)
        self.type = trader_type  # 1: Gösterge, 2: Sürü, 3: Aşırı güven, 4: Rastgele
        self.cash = 10000  # Başlangıç nakiti
        self.positions = 10  # Başlangıç pozisyon miktarı
        self.confidence = np.random.uniform(0.5, 1.0)  # Aşırı güven seviyesi (0.5-1.0)
        self.herd_tendency = np.random.uniform(0.3, 0.8) if trader_type == 2 else 0  # Sürü eğilimi
        self.trade_history = []
        self.last_action = 0 # last_action attribute'ünü ekledik

    def determine_trade_amount(self, price):
        """Güven seviyesine göre alım/satım miktarını belirle"""
        base_amount = 1  # Minimum işlem miktarı
        max_possible = min(self.cash / price if self.cash > 0 else 0,
                          self.positions if self.positions > 0 else 0)

        if self.type == 3:  # Aşırı güvenenler
            # Ensure max_possible is not zero before calculating
            if max_possible > 0:
                return min(base_amount + int(5 * self.confidence), max_possible)
            else:
                return 0 # Cannot trade if no cash or positions
        elif self.type == 2:  # Sürü davranışı
            # Ensure max_possible is not zero before calculating
            if max_possible > 0:
                return min(base_amount + int(3 * self.herd_tendency), max_possible)
            else:
                return 0 # Cannot trade if no cash or positions
        else: # Rastgele veya Gösterge
             if max_possible > 0:
                 return min(base_amount, max_possible)
             else:
                 return 0 # Cannot trade if no cash or positions

    def execute_trade(self, action, price):
        """Alım/satım işlemini gerçekleştir"""
        trade_amount = self.determine_trade_amount(price)

        if trade_amount == 0: # İşlem yapılamaz
            return 0

        if action == 1 and self.cash >= price * trade_amount:  # Alım
            self.positions += trade_amount
            self.cash -= price * trade_amount
            self.trade_history.append(('BUY', trade_amount, price))
            return trade_amount

        elif action == -1 and self.positions >= trade_amount:  # Satım
            self.positions -= trade_amount
            self.cash += price * trade_amount
            self.trade_history.append(('SELL', trade_amount, price))
            return -trade_amount

        return 0

    def step(self):
        """Güncellenmiş karar mekanizması"""
        price = self.model.current_price
        market_data = self.model.get_market_data()
        other_traders = self.model.agents # schedule.agents'i kullanarak tüm ajanlara erişim

        action = 0  # 0: Hold, 1: Buy, -1: Sell

        if self.type == 1:  # Gösterge temelli
            if market_data['rsi'] > 70 or market_data['news'] < 0:
                action = -1
            elif market_data['rsi'] < 30 or market_data['news'] > 0:
                action = 1

        elif self.type == 2:  # Sürü davranışı
            # Sürü davranışı için diğer ajanların son 5 eylemini alıyoruz.
            # Ancak agent listesi sırası garanti edilmediğinden son 5 rastgele olabilir.
            # Daha iyi bir model için ajanların eylemlerini merkezi olarak izlemek gerekebilir.
            # Basitlik için şimdilik son 5 ajanın eylemine bakıyoruz.
            recent_actions = [a.last_action for a in other_traders if hasattr(a, 'last_action')] # last_action attribute'ı olan ajanları filtrele

            if len(recent_actions) > 0: # Liste boş değilse devam et
                buy_count = sum(1 for a in recent_actions if a == 1)
                sell_count = sum(1 for a in recent_actions if a == -1)

                buy_ratio = buy_count / len(recent_actions)
                sell_ratio = sell_count / len(recent_actions)

                if buy_ratio > (0.5 + self.herd_tendency/2):
                    action = 1
                elif sell_ratio > (0.5 + self.herd_tendency/2): # Sürü sell eğilimi de ekleyelim
                    action = -1

        elif self.type == 3:  # Aşırı güven
             if len(self.trade_history) > 0:
                 last_trade = self.trade_history[-1]
                 # Kazanıp kazanmadığını kontrol et
                 current_value = self.positions * price + self.cash

                 # Çok basit bir kazanma/kaybetme kontrolü
                 # Gerçek kâr hesabı için daha detaylı trade geçmişi gerekir.
                 # Burada basitleştirip son trade'in yönüne ve fiyat değişimine bakıyoruz.
                 if (last_trade[0] == 'BUY' and price > last_trade[2]) or \
                    (last_trade[0] == 'SELL' and price < last_trade[2]):
                     # Kazanıyorsa daha fazla al/sat (aşırı güven)
                     action = 1 if last_trade[0] == 'BUY' else -1
                     action = action if np.random.random() < self.confidence else -action # Güvene göre hata payı

                 else:
                     # Kaybediyorsa rastgele veya ters işlem yapabilir (basitçe rastgele yapalım)
                      action = np.random.choice([-1, 0, 1]) # Güven seviyesi kayıpta nasıl etkiler? Burada basitleştirildi.
             else:
                 # İlk işlem rastgele güven seviyesine göre
                 action = 1 if np.random.random() < self.confidence else -1


        else:  # Rastgele
            action = np.random.choice([-1, 0, 1], p=[0.4, 0.2, 0.4])

        # İşlemi gerçekleştir
        trade_impact = self.execute_trade(action, price)
        self.last_action = action # Son eylemi güncelliyoruz

        # Model için işlem etkisini kaydet
        self.model.trade_impacts.append(trade_impact * price) # İşlem miktarının fiyatla çarpımı yerine sadece miktar daha mantıklı olabilir. Fiyat etkisi model adımında hesaplanmalı. Düzeltelim:

        self.model.trade_quantities.append(trade_impact) # İşlem miktarını kaydedelim


class MarketModel(mesa.Model):
    """Geliştirilmiş piyasa modeli"""
    def __init__(self, N=20, width=10, height=10):
        super().__init__()
        self.num_agents = N
        self.grid = MultiGrid(width, height, True)

        self.price_history = [100]
        self.trade_quantities = [] # İşlem miktarlarını tutacak liste
        self.current_step = 0

        # Ajanları oluştur (eşit dağılım)
        trader_types = [1, 2, 3, 4] * (N // 4)
        # Eğer N 4'e bölünemiyorsa kalan tipleri ekle
        remaining = N % 4
        if remaining > 0:
            trader_types.extend(random.sample([1, 2, 3, 4], remaining))

        random.shuffle(trader_types)

        for i in range(self.num_agents):
            # Agent Unique ID için model.next_id() kullanıyoruz
            trader = NoiseTrader(self, trader_types[i])
            self.agents.add(trader) # Ajanı schedule'a ekliyoruz

            # Rastgele bir hücreye yerleştir
            x, y = self.random.randrange(width), self.random.randrange(height)
            self.grid.place_agent(trader, (x, y))

        # Gelişmiş veri toplama
        self.datacollector = DataCollector(
            agent_reporters={
                "Cash": "cash",
                "Positions": "positions",
                "Type": "type",
                "Last_Action": "last_action" # last_action'ı da topluyoruz
            },
            model_reporters={
                "Price": "current_price",
                "Net_Demand": "current_net_demand", # Net talebi de topluyoruz
                "Total_Volume": "current_volume" # Toplam işlem hacmi
            }
        )

    def get_market_data(self):
        """Piyasa verilerini oluştur (şimdilik rastgele)"""
        # Gelecekte geçmiş fiyatlara dayalı RSI hesaplaması eklenebilir.
        # Haber etkisi daha sofistike modellenebilir.
        return {
            'rsi': np.random.uniform(20, 80), # Basit rastgele RSI
            'news': np.random.choice([-1, 0, 1], p=[0.1, 0.8, 0.1]) # Basit rastgele haber
        }

    @property
    def current_price(self):
        return self.price_history[-1]

    @property
    def current_net_demand(self):
        # Tüm ajanların işlem miktarlarının toplamı (pozitif alım, negatif satım)
        return sum(self.trade_quantities)

    @property
    def current_volume(self):
         # İşlem yapılan toplam miktar (mutlak değer)
         return sum(abs(q) for q in self.trade_quantities)

    def step(self):
        """Model adımı"""
        self.trade_quantities = []  # Yeni adım için işlem miktarları listesini temizle

        # Ajanların karar verip işlem yapması
        self.agents.do("step")

        # Fiyatı güncelle (net talep/arz etkisine göre)
        net_demand = self.current_net_demand
        # Fiyat değişimi = net talep * piyasa derinliği hassasiyeti
        # Negatif net talep fiyatı düşürür, pozitif net talep artırır.
        # Piyasa derinliği hassasiyeti (0.001 gibi küçük bir değer) talebin fiyata etkisini belirler.
        price_change_percentage = net_demand * 0.001 # Bu değer ayarlanabilir
        new_price = self.current_price * (1 + price_change_percentage)

        # Fiyat sınırlamaları (uç değerleri engellemek için)
        new_price = max(0.1, new_price) # Fiyatın 0 veya altına düşmesini engelle

        self.price_history.append(new_price)

        # Veri topla
        self.datacollector.collect(self)
        self.current_step += 1

def run_simulation(steps=200):
    """Simülasyonu çalıştır ve sonuçları göster"""
    model = MarketModel(N=20)

    for _ in range(steps):
        model.step()

    # Sonuçları görselleştir
    # DataCollector'dan verileri al
    model_vars = model.datacollector.get_model_vars_dataframe()
    agent_vars = model.datacollector.get_agent_vars_dataframe()


    fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(12, 15))

    # Fiyat grafiği
    ax1.plot(model_vars.index, model_vars['Price'])
    ax1.set_title("Piyasa Fiyat Hareketleri")
    ax1.set_xlabel("Zaman Adımı")
    ax1.set_ylabel("Fiyat")
    ax1.grid(True)

    # Toplam Net Talep grafiği
    ax2.plot(model_vars.index, model_vars['Net_Demand'], color='orange')
    ax2.set_title("Toplam Net Talep")
    ax2.set_xlabel("Zaman Adımı")
    ax2.set_ylabel("Net Talep Miktarı")
    ax2.grid(True)

    # Ajan varlıkları (son durum yerine zaman içindeki değişimi gösterelim)
    # Her bir ajan tipi için ortalama nakit ve pozisyonu gösterebiliriz.
    # Bu, çok sayıda ajan varsa daha anlamlı olur.
    # Tek tek ajanları göstermek isterseniz aşağıdaki kod kullanılabilir:
    # for agent_id in agent_vars.index.get_level_values('AgentID').unique():
    #     agent_data = agent_vars.xs(agent_id, level='AgentID')
    #     ax3.plot(agent_data.index, agent_data['Cash'], label=f'Agent {agent_id} (Cash)')
    #     ax3b = ax3.twinx()
    #     ax3b.plot(agent_data.index, agent_data['Positions'], label=f'Agent {agent_id} (Positions)', linestyle='--')

    # Ortalama varlıkları hesaplayalım
    agent_vars['Total_Value'] = agent_vars['Cash'] + agent_vars['Positions'] * model_vars['Price'].reindex(agent_vars.index, level='Step')

    avg_vars_by_type = agent_vars.groupby(['Step', 'Type']).mean().unstack()

    for trader_type in [1, 2, 3, 4]:
         if ('Total_Value', trader_type) in avg_vars_by_type.columns:
             ax3.plot(avg_vars_by_type.index, avg_vars_by_type[('Total_Value', trader_type)], label=f"Tip {trader_type} (Toplam Varlık)")

    ax3.set_title("Ajan Tiplerine Göre Ortalama Toplam Varlık Değişimi")
    ax3.set_xlabel("Zaman Adımı")
    ax3.set_ylabel("Ortalama Toplam Varlık (TL)")
    ax3.legend()
    ax3.grid(True)


    plt.tight_layout()
    plt.show()

    return model

if __name__ == "__main__":
    model = run_simulation(steps=200)


```

```python
#yukarıdaki hata veren kodun düzeltilmiş hali

import matplotlib.pyplot as plt
import numpy as np # numpy import edildi
import random # random import edildi

from mesa.space import MultiGrid
from mesa.datacollection import DataCollector


class NoiseTrader(mesa.Agent):
    """Geliştirilmiş gürültü tüccarı ajanı"""
    def __init__(self,model, trader_type):
        super().__init__(model)
        self.type = trader_type  # 1: Gösterge, 2: Sürü, 3: Aşırı güven, 4: Rastgele
        self.cash = 10000  # Başlangıç nakiti
        self.positions = 10  # Başlangıç pozisyon miktarı
        self.confidence = np.random.uniform(0.5, 1.0)  # Aşırı güven seviyesi (0.5-1.0)
        self.herd_tendency = np.random.uniform(0.3, 0.8) if trader_type == 2 else 0  # Sürü eğilimi
        self.trade_history = []
        self.last_action = 0 # last_action attribute'ünü ekledik

    def determine_trade_amount(self, price):
        """Güven seviyesine göre alım/satım miktarını belirle"""
        base_amount = 1  # Minimum işlem miktarı
        # Mevcut nakit ve pozisyonlarla yapılabilecek maksimum işlem miktarını hesapla
        max_buy_amount = self.cash / price if price > 0 and self.cash > 0 else 0
        max_sell_amount = self.positions if self.positions > 0 else 0

        if self.type == 3:  # Aşırı güvenenler
            # Aşırı güvenenler daha fazla işlem yapma eğiliminde
            intended_amount = base_amount + int(5 * self.confidence)
            return min(intended_amount, max_buy_amount) if self.last_action == 1 else min(intended_amount, max_sell_amount)

        elif self.type == 2:  # Sürü davranışı
            # Sürü davranışı olanlar da sürüye uyum oranında işlem yapma eğiliminde
            intended_amount = base_amount + int(3 * self.herd_tendency)
            return min(intended_amount, max_buy_amount) if self.last_action == 1 else min(intended_amount, max_sell_amount)

        else: # Rastgele veya Gösterge
             # Diğerleri temel işlem miktarını hedefler
             return min(base_amount, max_buy_amount) if self.last_action == 1 else min(base_amount, max_sell_amount)


    def execute_trade(self, action, price):
        """Alım/satım işlemini gerçekleştir"""
        trade_amount_to_execute = self.determine_trade_amount(price)

        if trade_amount_to_execute == 0: # İşlem yapılamaz
            return 0

        executed_amount = 0
        if action == 1 and self.cash >= price * trade_amount_to_execute:  # Alım
            executed_amount = trade_amount_to_execute
            self.positions += executed_amount
            self.cash -= price * executed_amount
            self.trade_history.append(('BUY', executed_amount, price))

        elif action == -1 and self.positions >= trade_amount_to_execute:  # Satım
            executed_amount = trade_amount_to_execute
            self.positions -= executed_amount
            self.cash += price * executed_amount
            self.trade_history.append(('SELL', executed_amount, price))
            executed_amount = -executed_amount # Satım miktarı negatif

        return executed_amount

    def step(self):
        """Güncellenmiş karar mekanizması"""
        price = self.model.current_price
        market_data = self.model.get_market_data()
        other_traders = self.model.agents # schedule.agents'i kullanarak tüm ajanlara erişim

        action = 0  # 0: Hold, 1: Buy, -1: Sell

        if self.type == 1:  # Gösterge temelli
            if market_data['rsi'] > 70 or market_data['news'] < 0:
                action = -1
            elif market_data['rsi'] < 30 or market_data['news'] > 0:
                action = 1

        elif self.type == 2:  # Sürü davranışı
            # Sürü davranışı için diğer ajanların son 5 eylemini alıyoruz.
            # Ancak agent listesi sırası garanti edilmediğinden son 5 rastgele olabilir.
            # Daha iyi bir model için ajanların eylemlerini merkezi olarak izlemek gerekebilir.
            # Basitlik için şimdilik tüm ajanların eylemlerine bakıyoruz.
            # Kendi action'ımızı hariç tutmak daha doğru olabilir, ancak bu simülasyon için şimdilik tümüne bakalım.
            recent_actions = [a.last_action for a in other_traders if hasattr(a, 'last_action')]

            if len(recent_actions) > 0:
                buy_count = sum(1 for a in recent_actions if a == 1)
                sell_count = sum(1 for a in recent_actions if a == -1)
                hold_count = len(recent_actions) - buy_count - sell_count

                # Basit sürü mekanizması: Çoğunluk ne yapıyorsa ona eğilim göster
                if buy_count > sell_count and buy_count > hold_count:
                     # Sürü alım yapıyor, ben de alım yapayım (sürü eğilimi oranında)
                     if np.random.random() < self.herd_tendency:
                         action = 1
                elif sell_count > buy_count and sell_count > hold_count:
                     # Sürü satım yapıyor, ben de satım yapayım (sürü eğilimi oranında)
                     if np.random.random() < self.herd_tendency:
                         action = -1
                # Eşitlik veya bekleme çoğunluktaysa action = 0 kalır


        elif self.type == 3:  # Aşırı güven
             if len(self.trade_history) > 0:
                 last_trade = self.trade_history[-1]
                 last_price = last_trade[2]
                 last_action_type = last_trade[0] # 'BUY' veya 'SELL'

                 # Kazanıp kazanmadığını kontrol et (basitçe son işlemden bu yana fiyatın yönü)
                 if (last_action_type == 'BUY' and price > last_price) or \
                    (last_action_type == 'SELL' and price < last_price):
                     # Kazanıyor: Aşırı güvenle aynı yönde işlem yapma eğilimi artar
                     if np.random.random() < self.confidence: # Güven seviyesi kadar aynı yönde işlem
                         action = 1 if last_action_type == 'BUY' else -1
                     else:
                          action = 0 # Kazanırken bile bazen bekleme

                 else: # Kaybediyor veya fiyat aynı
                     # Kaybederken veya fiyat aynıyken rastgele veya temkinli olabilir.
                     # Burada basitleştirip rastgele bir karar alalım (güven seviyesinin etkisi azalır)
                     if np.random.random() < (1 - (self.confidence - 0.5) * 2): # Güven arttıkça bu olasılık düşer
                         action = np.random.choice([-1, 0, 1])
                     else:
                         action = 0 # Daha temkinli bekleme

             else:
                 # İlk işlem aşırı güven seviyesine göre rastgele alım veya satım
                 action = 1 if np.random.random() < self.confidence else -1


        else:  # Rastgele
            action = np.random.choice([-1, 0, 1], p=[0.4, 0.2, 0.4]) # Satım, Bekleme, Alım olasılıkları

        # İşlemi gerçekleştir
        trade_quantity = self.execute_trade(action, price) # execute_trade artık miktarı döndürüyor
        self.last_action = action # Son eylemi güncelliyoruz

        # Model için işlem miktarını kaydet
        # Hata buradaydı. trade_impacts yerine trade_quantities kullanmalıyız.
        self.model.trade_quantities.append(trade_quantity)


class MarketModel(mesa.Model):
    """Geliştirilmiş piyasa modeli"""
    def __init__(self, N=20, width=10, height=10, seed=None):
        super().__init__()
        self.num_agents = N
        self.grid = MultiGrid(width, height, True)


        # <<<--- BURAYA SEED EKLEME KISMI --->>>
        if seed is not None:
            np.random.seed(seed) # Numpy'nin rastgele üretecini başlat
            random.seed(seed)   # Python'ın random üretecini başlat
            self.random.seed(seed) # Mesa modelinin kendi üretecini başlat
        # <<<--------------------------------->>>


        self.price_history = [100]
        self.trade_quantities = [] # İşlem miktarlarını tutacak liste
        self.current_step = 0

        # Ajanları oluştur (eşit dağılım)
        trader_types = [1, 2, 3, 4] * (N // 4)
        # Eğer N 4'e bölünemiyorsa kalan tipleri ekle
        remaining = N % 4
        if remaining > 0:
            trader_types.extend(random.sample([1, 2, 3, 4], remaining))

        random.shuffle(trader_types)

        for i in range(self.num_agents):
            # Agent Unique ID için model.next_id() kullanıyoruz
            trader = NoiseTrader(self, trader_types[i])
            self.agents.add(trader) # Ajanı schedule'a ekliyoruz

            # Rastgele bir hücreye yerleştir (grid kullanımı zorunlu değil ama modelde tanımlı)
            x, y = self.random.randrange(width), self.random.randrange(height)
            self.grid.place_agent(trader, (x, y))


        # Gelişmiş veri toplama
        self.datacollector = DataCollector(
            agent_reporters={
                "Cash": "cash",
                "Positions": "positions",
                "Type": "type",
                "Last_Action": "last_action" # last_action'ı da topluyoruz
            },
            model_reporters={
                "Price": "current_price",
                "Net_Demand": "current_net_demand", # Net talebi de topluyoruz
                "Total_Volume": "current_volume" # Toplam işlem hacmi
            }
        )

    def get_market_data(self):
        """Piyasa verilerini oluştur (şimdilik rastgele)"""
        # Gelecekte geçmiş fiyatlara dayalı RSI hesaplaması eklenebilir.
        # Haber etkisi daha sofistike modellenebilir.
        return {
            'rsi': np.random.uniform(20, 80), # Basit rastgele RSI
            'news': np.random.choice([-1, 0, 1], p=[0.1, 0.8, 0.1]) # Basit rastgele haber
        }

    @property
    def current_price(self):
        return self.price_history[-1]

    @property
    def current_net_demand(self):
        # Tüm ajanların işlem miktarlarının toplamı (pozitif alım, negatif satım)
        return sum(self.trade_quantities)

    @property
    def current_volume(self):
         # İşlem yapılan toplam miktar (mutlak değer)
         return sum(abs(q) for q in self.trade_quantities)

    def step(self):
        """Model adımı"""
        self.trade_quantities = []  # Yeni adım için işlem miktarları listesini temizle

        # Ajanların karar verip işlem yapması
        # Mesa'nın schedule'ını kullanarak ajanların step metodunu çalıştırıyoruz
        self.agents.do("step")

        # Fiyatı güncelle (net talep/arz etkisine göre)
        net_demand = self.current_net_demand
        # Fiyat değişimi = net talep * piyasa derinliği hassasiyeti
        # Negatif net talep fiyatı düşürür, pozitif net talep artırır.
        # Piyasa derinliği hassasiyeti (0.001 gibi küçük bir değer) talebin fiyata etkisini belirler.
        price_change_percentage = net_demand * 0.001 # Bu değer ayarlanabilir
        new_price = self.current_price * (1 + price_change_percentage)

        # Fiyat sınırlamaları (uç değerleri engellemek için)
        new_price = max(0.1, new_price) # Fiyatın 0 veya altına düşmesini engelle

        self.price_history.append(new_price)

        # Veri topla
        self.datacollector.collect(self)
        self.current_step += 1

def run_simulation(steps=200, seed=None):
    """Simülasyonu çalıştır ve sonuçları göster"""
    model = MarketModel(N=20, seed=seed)

    for _ in range(steps):
        model.step()

    # Sonuçları görselleştir
    # DataCollector'dan verileri al
    model_vars = model.datacollector.get_model_vars_dataframe()
    agent_vars = model.datacollector.get_agent_vars_dataframe()


    fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(12, 15))

    # Fiyat grafiği
    ax1.plot(model_vars.index, model_vars['Price'])
    ax1.set_title("Piyasa Fiyat Hareketleri")
    ax1.set_xlabel("Zaman Adımı")
    ax1.set_ylabel("Fiyat")
    ax1.grid(True)

    # Toplam Net Talep grafiği
    ax2.plot(model_vars.index, model_vars['Net_Demand'], color='orange')
    ax2.set_title("Toplam Net Talep")
    ax2.set_xlabel("Zaman Adımı")
    ax2.set_ylabel("Net Talep Miktarı")
    ax2.grid(True)

    # Ajan varlıkları (son durum yerine zaman içindeki değişimi gösterelim)
    # Her bir ajan tipi için ortalama nakit ve pozisyonu gösterebiliriz.
    # Bu, çok sayıda ajan varsa daha anlamlı olur.
    # Tek tek ajanları göstermek isterseniz aşağıdaki kod kullanılabilir:
    # for agent_id in agent_vars.index.get_level_values('AgentID').unique():
    #     agent_data = agent_vars.xs(agent_id, level='AgentID')
    #     ax3.plot(agent_data.index, agent_data['Cash'], label=f'Agent {agent_id} (Cash)')
    #     ax3b = ax3.twinx()
    #     ax3b.plot(agent_data.index, agent_data['Positions'], label=f'Agent {agent_id} (Positions)', linestyle='--')

    # Ortalama varlıkları hesaplayalım
    # agent_vars ve model_vars'ın indexlerini eşleştirmek için reindex kullan
    # Model veri DataFrame'inin indexini MultiIndex'e (Step, AgentID) genişlet
    price_series = model_vars['Price'].reindex(agent_vars.index, level='Step')

    # Total_Value hesaplaması yaparken NaN değerleri kontrol et
    # Fiyat NaN ise Total_Value hesaplanamaz, bu durumda 0 veya başka bir değer atayabiliriz.
    agent_vars['Total_Value'] = agent_vars['Cash'] + agent_vars['Positions'] * price_series
    agent_vars['Total_Value'].fillna(agent_vars['Cash'], inplace=True) # Fiyat NaN ise sadece nakiti kullan

    avg_vars_by_type = agent_vars.groupby(['Step', 'Type']).mean().unstack()

    for trader_type in [1, 2, 3, 4]:
         if ('Total_Value', trader_type) in avg_vars_by_type.columns:
             ax3.plot(avg_vars_by_type.index, avg_vars_by_type[('Total_Value', trader_type)], label=f"Tip {trader_type} (Ortalama Toplam Varlık)")

    ax3.set_title("Ajan Tiplerine Göre Ortalama Toplam Varlık Değişimi")
    ax3.set_xlabel("Zaman Adımı")
    ax3.set_ylabel("Ortalama Toplam Varlık (TL)")
    ax3.legend()
    ax3.grid(True)


    plt.tight_layout()
    plt.show()

    return model

if __name__ == "__main__":
    model = run_simulation(steps=200, seed=42)
```

**yukarıdaki kodda ajanların nakit ve varlık miktarları aynı. Eğer Nakit ve miktarları faklı olsaydı nasıl olurdu?**

```python
#yukarıdaki hata veren kodun düzeltilmiş hali

import matplotlib.pyplot as plt
import numpy as np # numpy import edildi
import random # random import edildi

from mesa.space import MultiGrid
from mesa.datacollection import DataCollector


class NoiseTrader(mesa.Agent):
    """Geliştirilmiş gürültü tüccarı ajanı"""
    def __init__(self,model, trader_type):
        super().__init__(model)
        self.type = trader_type  # 1: Gösterge, 2: Sürü, 3: Aşırı güven, 4: Rastgele
        self.cash = np.random.uniform(10000, 100000)   # Başlangıç nakiti
        self.positions = np.random.uniform(10, 100)  # Başlangıç pozisyon miktarı
        self.confidence = np.random.uniform(0.5, 1.0)  # Aşırı güven seviyesi (0.5-1.0)
        self.herd_tendency = np.random.uniform(0.3, 0.8) if trader_type == 2 else 0  # Sürü eğilimi
        self.trade_history = []
        self.last_action = 0 # last_action attribute'ünü ekledik

    def determine_trade_amount(self, price):
        """Güven seviyesine göre alım/satım miktarını belirle"""
        base_amount = 1  # Minimum işlem miktarı
        # Mevcut nakit ve pozisyonlarla yapılabilecek maksimum işlem miktarını hesapla
        max_buy_amount = self.cash / price if price > 0 and self.cash > 0 else 0
        max_sell_amount = self.positions if self.positions > 0 else 0

        if self.type == 3:  # Aşırı güvenenler
            # Aşırı güvenenler daha fazla işlem yapma eğiliminde
            intended_amount = base_amount + int(5 * self.confidence)
            return min(intended_amount, max_buy_amount) if self.last_action == 1 else min(intended_amount, max_sell_amount)

        elif self.type == 2:  # Sürü davranışı
            # Sürü davranışı olanlar da sürüye uyum oranında işlem yapma eğiliminde
            intended_amount = base_amount + int(3 * self.herd_tendency)
            return min(intended_amount, max_buy_amount) if self.last_action == 1 else min(intended_amount, max_sell_amount)

        else: # Rastgele veya Gösterge
             # Diğerleri temel işlem miktarını hedefler
             return min(base_amount, max_buy_amount) if self.last_action == 1 else min(base_amount, max_sell_amount)


    def execute_trade(self, action, price):
        """Alım/satım işlemini gerçekleştir"""
        trade_amount_to_execute = self.determine_trade_amount(price)

        if trade_amount_to_execute == 0: # İşlem yapılamaz
            return 0

        executed_amount = 0
        if action == 1 and self.cash >= price * trade_amount_to_execute:  # Alım
            executed_amount = trade_amount_to_execute
            self.positions += executed_amount
            self.cash -= price * executed_amount
            self.trade_history.append(('BUY', executed_amount, price))

        elif action == -1 and self.positions >= trade_amount_to_execute:  # Satım
            executed_amount = trade_amount_to_execute
            self.positions -= executed_amount
            self.cash += price * executed_amount
            self.trade_history.append(('SELL', executed_amount, price))
            executed_amount = -executed_amount # Satım miktarı negatif

        return executed_amount

    def step(self):
        """Güncellenmiş karar mekanizması"""
        price = self.model.current_price
        market_data = self.model.get_market_data()
        other_traders = self.model.agents # schedule.agents'i kullanarak tüm ajanlara erişim

        action = 0  # 0: Hold, 1: Buy, -1: Sell

        if self.type == 1:  # Gösterge temelli
            if market_data['rsi'] > 70 or market_data['news'] < 0:
                action = -1
            elif market_data['rsi'] < 30 or market_data['news'] > 0:
                action = 1

        elif self.type == 2:  # Sürü davranışı
            # Sürü davranışı için diğer ajanların son 5 eylemini alıyoruz.
            # Ancak agent listesi sırası garanti edilmediğinden son 5 rastgele olabilir.
            # Daha iyi bir model için ajanların eylemlerini merkezi olarak izlemek gerekebilir.
            # Basitlik için şimdilik tüm ajanların eylemlerine bakıyoruz.
            # Kendi action'ımızı hariç tutmak daha doğru olabilir, ancak bu simülasyon için şimdilik tümüne bakalım.
            recent_actions = [a.last_action for a in other_traders if hasattr(a, 'last_action')]

            if len(recent_actions) > 0:
                buy_count = sum(1 for a in recent_actions if a == 1)
                sell_count = sum(1 for a in recent_actions if a == -1)
                hold_count = len(recent_actions) - buy_count - sell_count

                # Basit sürü mekanizması: Çoğunluk ne yapıyorsa ona eğilim göster
                if buy_count > sell_count and buy_count > hold_count:
                     # Sürü alım yapıyor, ben de alım yapayım (sürü eğilimi oranında)
                     if np.random.random() < self.herd_tendency:
                         action = 1
                elif sell_count > buy_count and sell_count > hold_count:
                     # Sürü satım yapıyor, ben de satım yapayım (sürü eğilimi oranında)
                     if np.random.random() < self.herd_tendency:
                         action = -1
                # Eşitlik veya bekleme çoğunluktaysa action = 0 kalır


        elif self.type == 3:  # Aşırı güven
             if len(self.trade_history) > 0:
                 last_trade = self.trade_history[-1]
                 last_price = last_trade[2]
                 last_action_type = last_trade[0] # 'BUY' veya 'SELL'

                 # Kazanıp kazanmadığını kontrol et (basitçe son işlemden bu yana fiyatın yönü)
                 if (last_action_type == 'BUY' and price > last_price) or \
                    (last_action_type == 'SELL' and price < last_price):
                     # Kazanıyor: Aşırı güvenle aynı yönde işlem yapma eğilimi artar
                     if np.random.random() < self.confidence: # Güven seviyesi kadar aynı yönde işlem
                         action = 1 if last_action_type == 'BUY' else -1
                     else:
                          action = 0 # Kazanırken bile bazen bekleme

                 else: # Kaybediyor veya fiyat aynı
                     # Kaybederken veya fiyat aynıyken rastgele veya temkinli olabilir.
                     # Burada basitleştirip rastgele bir karar alalım (güven seviyesinin etkisi azalır)
                     if np.random.random() < (1 - (self.confidence - 0.5) * 2): # Güven arttıkça bu olasılık düşer
                         action = np.random.choice([-1, 0, 1])
                     else:
                         action = 0 # Daha temkinli bekleme

             else:
                 # İlk işlem aşırı güven seviyesine göre rastgele alım veya satım
                 action = 1 if np.random.random() < self.confidence else -1


        else:  # Rastgele
            action = np.random.choice([-1, 0, 1], p=[0.4, 0.2, 0.4]) # Satım, Bekleme, Alım olasılıkları

        # İşlemi gerçekleştir
        trade_quantity = self.execute_trade(action, price) # execute_trade artık miktarı döndürüyor
        self.last_action = action # Son eylemi güncelliyoruz

        # Model için işlem miktarını kaydet
        # Hata buradaydı. trade_impacts yerine trade_quantities kullanmalıyız.
        self.model.trade_quantities.append(trade_quantity)


class MarketModel(mesa.Model):
    """Geliştirilmiş piyasa modeli"""
    def __init__(self, N=20, width=10, height=10, seed=None):
        super().__init__()
        self.num_agents = N
        self.grid = MultiGrid(width, height, True)

        # <<<--- BURAYA SEED EKLEME KISMI --->>>
        if seed is not None:
            np.random.seed(seed) # Numpy'nin rastgele üretecini başlat
            random.seed(seed)   # Python'ın random üretecini başlat
            self.random.seed(seed) # Mesa modelinin kendi üretecini başlat
        # <<<--------------------------------->>>


        self.price_history = [100]
        self.trade_quantities = [] # İşlem miktarlarını tutacak liste
        self.current_step = 0

        # Ajanları oluştur (eşit dağılım)
        trader_types = [1, 2, 3, 4] * (N // 4)
        # Eğer N 4'e bölünemiyorsa kalan tipleri ekle
        remaining = N % 4
        if remaining > 0:
            trader_types.extend(random.sample([1, 2, 3, 4], remaining))

        random.shuffle(trader_types)

        for i in range(self.num_agents):
            # Agent Unique ID için model.next_id() kullanıyoruz
            trader = NoiseTrader(self, trader_types[i])
            self.agents.add(trader) # Ajanı schedule'a ekliyoruz

            # Rastgele bir hücreye yerleştir (grid kullanımı zorunlu değil ama modelde tanımlı)
            x, y = self.random.randrange(width), self.random.randrange(height)
            self.grid.place_agent(trader, (x, y))


        # Gelişmiş veri toplama
        self.datacollector = DataCollector(
            agent_reporters={
                "Cash": "cash",
                "Positions": "positions",
                "Type": "type",
                "Last_Action": "last_action" # last_action'ı da topluyoruz
            },
            model_reporters={
                "Price": "current_price",
                "Net_Demand": "current_net_demand", # Net talebi de topluyoruz
                "Total_Volume": "current_volume" # Toplam işlem hacmi
            }
        )

    def get_market_data(self):
        """Piyasa verilerini oluştur (şimdilik rastgele)"""
        # Gelecekte geçmiş fiyatlara dayalı RSI hesaplaması eklenebilir.
        # Haber etkisi daha sofistike modellenebilir.
        return {
            'rsi': np.random.uniform(20, 80), # Basit rastgele RSI
            'news': np.random.choice([-1, 0, 1], p=[0.1, 0.8, 0.1]) # Basit rastgele haber
        }

    @property
    def current_price(self):
        return self.price_history[-1]

    @property
    def current_net_demand(self):
        # Tüm ajanların işlem miktarlarının toplamı (pozitif alım, negatif satım)
        return sum(self.trade_quantities)

    @property
    def current_volume(self):
         # İşlem yapılan toplam miktar (mutlak değer)
         return sum(abs(q) for q in self.trade_quantities)

    def step(self):
        """Model adımı"""
        self.trade_quantities = []  # Yeni adım için işlem miktarları listesini temizle

        # Ajanların karar verip işlem yapması
        # Mesa'nın schedule'ını kullanarak ajanların step metodunu çalıştırıyoruz
        self.agents.do("step")

        # Fiyatı güncelle (net talep/arz etkisine göre)
        net_demand = self.current_net_demand
        # Fiyat değişimi = net talep * piyasa derinliği hassasiyeti
        # Negatif net talep fiyatı düşürür, pozitif net talep artırır.
        # Piyasa derinliği hassasiyeti (0.001 gibi küçük bir değer) talebin fiyata etkisini belirler.
        price_change_percentage = net_demand * 0.001 # Bu değer ayarlanabilir
        new_price = self.current_price * (1 + price_change_percentage)

        # Fiyat sınırlamaları (uç değerleri engellemek için)
        new_price = max(0.1, new_price) # Fiyatın 0 veya altına düşmesini engelle

        self.price_history.append(new_price)

        # Veri topla
        self.datacollector.collect(self)
        self.current_step += 1

def run_simulation(steps=200, seed=None):
    """Simülasyonu çalıştır ve sonuçları göster"""
    model = MarketModel(N=20, seed=seed)

    for _ in range(steps):
        model.step()

    # Sonuçları görselleştir
    # DataCollector'dan verileri al
    model_vars = model.datacollector.get_model_vars_dataframe()
    agent_vars = model.datacollector.get_agent_vars_dataframe()


    fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(12, 15))

    # Fiyat grafiği
    ax1.plot(model_vars.index, model_vars['Price'])
    ax1.set_title("Piyasa Fiyat Hareketleri")
    ax1.set_xlabel("Zaman Adımı")
    ax1.set_ylabel("Fiyat")
    ax1.grid(True)

    # Toplam Net Talep grafiği
    ax2.plot(model_vars.index, model_vars['Net_Demand'], color='orange')
    ax2.set_title("Toplam Net Talep")
    ax2.set_xlabel("Zaman Adımı")
    ax2.set_ylabel("Net Talep Miktarı")
    ax2.grid(True)

    # Ajan varlıkları (son durum yerine zaman içindeki değişimi gösterelim)
    # Her bir ajan tipi için ortalama nakit ve pozisyonu gösterebiliriz.
    # Bu, çok sayıda ajan varsa daha anlamlı olur.
    # Tek tek ajanları göstermek isterseniz aşağıdaki kod kullanılabilir:
    # for agent_id in agent_vars.index.get_level_values('AgentID').unique():
    #     agent_data = agent_vars.xs(agent_id, level='AgentID')
    #     ax3.plot(agent_data.index, agent_data['Cash'], label=f'Agent {agent_id} (Cash)')
    #     ax3b = ax3.twinx()
    #     ax3b.plot(agent_data.index, agent_data['Positions'], label=f'Agent {agent_id} (Positions)', linestyle='--')

    # Ortalama varlıkları hesaplayalım
    # agent_vars ve model_vars'ın indexlerini eşleştirmek için reindex kullan
    # Model veri DataFrame'inin indexini MultiIndex'e (Step, AgentID) genişlet
    price_series = model_vars['Price'].reindex(agent_vars.index, level='Step')

    # Total_Value hesaplaması yaparken NaN değerleri kontrol et
    # Fiyat NaN ise Total_Value hesaplanamaz, bu durumda 0 veya başka bir değer atayabiliriz.
    agent_vars['Total_Value'] = agent_vars['Cash'] + agent_vars['Positions'] * price_series
    agent_vars['Total_Value'].fillna(agent_vars['Cash'], inplace=True) # Fiyat NaN ise sadece nakiti kullan

    avg_vars_by_type = agent_vars.groupby(['Step', 'Type']).mean().unstack()

    for trader_type in [1, 2, 3, 4]:
         if ('Total_Value', trader_type) in avg_vars_by_type.columns:
             ax3.plot(avg_vars_by_type.index, avg_vars_by_type[('Total_Value', trader_type)], label=f"Tip {trader_type} (Ortalama Toplam Varlık)")

    ax3.set_title("Ajan Tiplerine Göre Ortalama Toplam Varlık Değişimi")
    ax3.set_xlabel("Zaman Adımı")
    ax3.set_ylabel("Ortalama Toplam Varlık (TL)")
    ax3.legend()
    ax3.grid(True)


    plt.tight_layout()
    plt.show()

    return model

if __name__ == "__main__":
    model = run_simulation(steps=200, seed=42)
```

Elbette, sağlanan en son kod hücresindeki Python kodunu algoritmik olarak açıklayalım. Kod, finansal piyasalarda "gürültü tüccarları"nın (noise traders) davranışlarını modelleyen bir ajan tabanlı simülasyondur.

Genel Algoritma:

Simülasyon, bir finansal piyasayı ve bu piyasada işlem yapan farklı türdeki gürültü tüccarlarını modeller. Her simülasyon adımında (step), ajanlar mevcut piyasa durumuna (fiyat, haber, göstergeler) ve kendi içsel özelliklerine (tür, güven, sürü eğilimi) göre alım veya satım kararı verirler. Bu işlemlerin toplamı, piyasa fiyatını etkiler ve yeni fiyat bir sonraki adım için temel oluşturur. Simülasyon belirli bir adım sayısı boyunca çalışır ve sonunda piyasa fiyatının ve ajanların portföylerinin zaman içindeki değişimi görselleştirilir.

Temel Bileşenler ve Algoritmaları:

NoiseTrader Sınıfı (Ajan):

    __init__(self, unique_id, model, trader_type):

        Bir tüccar ajanı oluşturulduğunda çalışır.
        Ajana benzersiz bir kimlik (unique_id) atar.

        Ajanın ait olduğu model örneğine referans saklar.

        Ajanın türünü belirler (trader_type: 1-Gösterge, 2-Sürü, 3-Aşırı Güven, 4-Rastgele).

        Başlangıç nakit ve pozisyon (varlık) miktarını ayarlar.

        Türe göre aşırı güven seviyesi veya sürü eğilimi gibi davranışsal parametreleri rastgele belirler.

        Boş bir işlem geçmişi (trade_history) listesi başlatır.

        Son yaptığı işlemi (last_action) saklamak için bir öznitelik başlatır.

    determine_trade_amount(self, price):

        Ajana göre işlem yapmayı düşündüğü miktarı belirler.

        Minimum işlem miktarı (base_amount) tanımlar (genellikle 1).

        Mevcut nakit ve pozisyonlara göre yapabileceği maksimum alım miktarını (nakit / fiyat) ve maksimum satım miktarını (pozisyonlar) hesaplar.

        Ajanın türüne göre, base_amount'a davranışsal parametrelerin (güven, sürü eğilimi) etkisini ekleyerek niyetlenen işlem miktarını hesaplar.

        Hesaplanan niyetlenen miktarı, last_actiona göre ilgili maksimum miktar (alım veya satım) ile karşılaştırır ve hangisi daha küçükse o miktarı döndürür. Bu, ajanın yapabileceğinden fazla işlem talep etmesini engeller.

    execute_trade(self, action, price): Belirlenen işi (action) gerçekleştirir.

        determine_trade_amount metodunu çağırarak ne kadar işlem yapacağını belirler.

        Eğer belirlenen miktar 0 ise veya action 0 (bekle) ise işlem yapmaz ve 0 döndürür.

        Eğer action 1 (alım) ise VE yeterli nakit varsa:

            Pozisyonları artırır, nakiti azaltır.

            İşlemi trade_history'ye kaydeder.

            Gerçekleşen pozitif işlem miktarını döndürür.

        Eğer action -1 (satım) ise VE yeterli pozisyon varsa:

            Pozisyonları azaltır, nakiti artırır.

            İşlemi trade_history'ye kaydeder.

            Gerçekleşen negatif işlem miktarını döndürür.

    step(self): Ajanın her simülasyon adımında çalışacak ana mantık.

        Modelden güncel piyasa fiyatını alır.

        Modelden güncel piyasa verilerini (rastgele RSI ve haber) alır.

        Modelin schedule'ından diğer tüm ajanlara erişir.

        Ajanın türüne göre bir action (1: Al, -1: Sat, 0: Bekle) kararı verir:

            Tip 1 (Gösterge): RSI ve haber verilerine bakarak eşik değerlere göre karar verir.

            Tip 2 (Sürü): Diğer ajanların last_action bilgilerine bakarak, çoğunluğun hareketini kendi herd_tendency (sürü eğilimi) oranı kadar takip etme olasılığına göre karar verir.

            Tip 3 (Aşırı Güven): En son işlemine ve mevcut fiyata bakarak kazanıp kazanmadığını kontrol eder. Kazanıyorsa, kendi confidence (güven) seviyesine göre aynı yönde daha fazla işlem yapma eğilimindedir. Kaybediyorsa, rastgele veya temkinli bir karar alır.

            Tip 4 (Rastgele): Belirlenen olasılıklarla (-1, 0, 1) rastgele bir karar seçer.

        Aldığı action kararını ve güncel price bilgisini kullanarak execute_trade metodunu çağırır.

        execute_trade metodunun döndürdüğü gerçekleşen işlem miktarını (trade_quantity) alır.

        Bu adımdaki action kararını self.last_action özniteliğine kaydeder.

        Gerçekleşen trade_quantity'yi modelin trade_quantities listesine ekler. Bu, ajanın piyasaya etkisini modele iletmek içindir.

MarketModel Sınıfı (Model):

__init__(self, N=20, width=10, height=10, seed=None): Simülasyon modelini başlatır.

    Simülasyondaki ajan sayısını (N) belirler.

    Ajanların yerleşeceği bir grid (ızgara) alanı oluşturur (konumsal modelleme için ama bu modelde konumsal etkileşim kullanılmıyor).

    Mesa'nın schedule (zamanlama) mekanizmasını başlatır (RandomActivation: her adımda ajanlar rastgele sırada adım atar).

    Bir seed değeri verildiyse, numpy, random ve self.random rastgele sayı üreteçlerini bu seed ile başlatır (tekrar üretilebilirlik için).

    Başlangıç fiyatı ile bir price_history listesi başlatır.

    Her adımdaki toplam işlem miktarlarını toplamak için boş bir trade_quantities listesi başlatır.
    self.current_step sayacını başlatır.

    Belirlenen N sayısına göre farklı türlerde ajanlar oluşturur (türler eşit olarak dağıtılmaya çalışılır), her ajana benzersiz kimlik atar ve onları schedule'a ekler.

    Ajanları grid üzerine rastgele yerleştirir (bu modelde grid kullanılmasa da).

    Simülasyon boyunca veri toplamak için bir DataCollector başlatır. Hangi ajan ve model değişkenlerinin toplanacağını belirler.

get_market_data(self): Piyasa için rastgele veri üretir.

    Bu modelde, rastgele bir RSI değeri (20-80 arası) ve rastgele bir haber değeri (-1: kötü, 0: nötr, 1: iyi) döndürür. (Bu, gerçek piyasa göstergelerinin ve haberlerinin çok basitleştirilmiş bir modelidir.)

current_price (property): price_history listesindeki son elemanı (mevcut fiyatı) döndüren bir özelliktir.

current_net_demand (property): trade_quantities listesindeki tüm işlem miktarlarının toplamını döndüren bir özelliktir. Bu, alım (pozitif) ve satım (negatif) işlemlerinin dengesini gösterir.

current_volume (property): trade_quantities listesindeki tüm işlem miktarlarının mutlak değerlerinin toplamını döndüren bir özelliktir. Bu, o adımdaki toplam işlem hacmini gösterir.

step(self): Modelin her simülasyon adımında çalışacak ana mantık.

    trade_quantities listesini temizler (bu adım için).

    self.schedule.step(): Schedule'daki tüm ajanların step metodunu sırayla (RandomActivation kullanılıyorsa rastgele sırada) çalıştırır. Her ajanın step metodu çalışırken, o ajanın yaptığı işlemlerin miktarı modelin trade_quantities listesine eklenir.

    Ajanların tüm adımları tamamlandıktan sonra, modelin current_net_demand özelliğini kullanarak toplam net talebi hesaplar.

    Net talebe göre fiyat değişim yüzdesini hesaplar (net talep * piyasa derinliği hassasiyeti). Pozitif net talep fiyatı artırır, negatif net talep düşürür.

    Mevcut fiyata göre yeni fiyatı hesaplar.

    Yeni fiyatın çok düşük veya çok yüksek olmasını engelleyen sınırlamalar uygular (örn: fiyat 0.1'in altına düşemez).

    Yeni fiyatı price_history listesine ekler.

    DataCollector'ı çalıştırarak o adımdaki model ve ajan verilerini toplar.

    self.current_step sayacını bir artırır.

run_simulation(steps=200, seed=None) Fonksiyonu:

    Simülasyonu başlatmak ve çalıştırmak için kullanılır.

    Kaç adım (steps) süreceğini ve hangi seed değerinin kullanılacağını belirler.

    Belirtilen seed ile bir MarketModel örneği oluşturur.

    Belirtilen steps sayısı kadar modelin step() metodunu bir döngü içinde çalıştırır.

    Simülasyon tamamlandıktan sonra, DataCollector'dan toplanan verileri alır (model ve ajan değişkenleri).

    Sonuçları görselleştirir:

          Piyasa fiyatının zaman içindeki değişimini bir çizgi grafik ile gösterir.

          Toplam net talebin zaman içindeki değişimini bir çizgi grafik ile gösterir.

          Her ajan türünün ortalama toplam varlığının (nakit + pozisyonlar * fiyat) zaman içindeki değişimini bir çizgi grafik ile gösterir. Bu, farklı stratejilere sahip ajanların simülasyon sonunda ne kadar başarılı olduğunu anlamaya yardımcı olur.

    Grafikleri gösterir (plt.show()).

    Çalışan model örneğini döndürür (analiz için kullanılabilir).

Algoritmik Akış (Bir Simülasyon Adımı):

    Model, o adımdaki toplam işlem miktarları listesini (trade_quantities) sıfırlar.

    Modelin schedule'ı, tüm ajanların listesini alır ve onları rastgele (RandomActivation kullanıldığı için) karıştırır.

    Karıştırılmış sıraya göre her ajan için sırasıyla şu adımlar çalışır: a. Ajan, modelden güncel piyasa fiyatını ve rastgele piyasa verilerini alır. b. Ajan, kendi türüne ve bu bilgilere (ve diğer ajanların son eylemlerine bakarak) dayanarak bir alım, satım veya bekleme kararı (action) verir. c. Ajan, determine_trade_amount ile ne kadar işlem yapmak istediğini belirler. d. Ajan, execute_trade ile kararını ve miktarı uygulamaya çalışır. Yeterli kaynağı varsa işlem gerçekleşir, portföyü güncellenir ve gerçekleşen miktar döndürülür. e. Ajan, o adımdaki kararını (action) kaydeder. f. Ajan, gerçekleşen işlem miktarını modelin trade_quantities listesine ekler.

    Tüm ajanlar adımlarını tamamladıktan sonra: a. Model, ajanların eklediği miktarların toplamından net talebi hesaplar. b. Net talebi kullanarak fiyat değişimini hesaplar ve yeni fiyatı belirler. c. Yeni fiyatı fiyat geçmişine ekler. d. Model, o adımdaki model değişkenlerini ve ajan değişkenlerini (nakit, pozisyon, tür, son aksiyon) DataCollector'a kaydeder.

    Simülasyon süresi dolana kadar bu adımlar tekrarlanır.

    Simülasyon bittikten sonra toplanan veriler analiz edilir ve görselleştirilir.

Bu yapı, farklı gürültü tüccarı türlerinin davranışlarının birleşerek piyasa fiyatını nasıl etkilediğini dinamik olarak gözlemlemeye olanak tanır.
