YUKARIDAKİ BAYES KODU İLE AŞAĞIDAKİ BAYES KODU ARASINDAKİ FARK NEDİR:

İki Bayesian yaklaşımı arasındaki farkları daha detaylı bir şekilde açıklayayım:

İlk Bayesian Yaklaşımı (YUKARIDAKİ)

Bu yaklaşım, piyasa senaryoları (boğa, ayı, aralık) hakkındaki inançları yeni verilerle güncellemeye odaklanır. Özeti şöyledir:

Önsel (Prior): Her piyasa senaryosunun olasılığı hakkında bir başlangıç inancıyla (önsel olasılıklar) başlar.

Olabilirlik (Likelihood): Her senaryo göz önüne alındığında, yeni verilerin gözlemlenme olasılığını hesaplar.

Sonrasal (Posterior): Senaryolar üzerinde bir sonrasal olasılık dağılımı elde etmek için Bayes teoremini kullanarak önsel inancı günceller.

Optimal Strateji: Senaryoların güncellenmiş sonrasal olasılıklarına göre en yüksek beklenen getiriyi sağlayan stratejiyi seçer.

İkinci Bayesian Yaklaşımı (AŞAĞIDAKİ)

"Bayesian Karar Teorisi" olarak adlandırılan bu yaklaşım, biraz farklı bir bakış açısına sahiptir. Her stratejinin beklenen faydasını doğrudan hesaplamaya ve faydayı en üst düzeye çıkaranı seçmeye odaklanır.

Sonrasal Olasılıklar (Dirichlet): Senaryolar hakkındaki önsel inançları açıkça güncellemek yerine, stratejiler üzerinde bir olasılık dağılımı oluşturmak için bir Dirichlet dağılımı kullanır (posterior_probs). Bu dağılım, hangi stratejinin en iyi olduğu konusundaki belirsizliği temsil eder.
Fayda Fonksiyonu: Farklı sonuçların (bu durumda ortalama getiriler) arzu edilirliğini nicelleştiren bir fayda fonksiyonu (U = np.log(1 + avg_return)) tanımlar. Logaritmik form genellikle riskten kaçınmayı temsil etmek için kullanılır.
Beklenen Fayda: Her strateji için, stratejinin sonrasal olasılığını faydası ile çarparak beklenen faydayı hesaplar.
Optimal Strateji: En yüksek beklenen faydaya sahip stratejiyi seçer (chosen_bayesian).
