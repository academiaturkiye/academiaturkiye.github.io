3. Simülasyon Modeli İçin Ne Anlama Gelir?
3.1. Ajanlar
Her ajan, haber akışını değerlendirir.

Ajanın haber okuma hassasiyeti, algıladığı haber etkisi ve risk profili farklı olabilir.

Evrimsel güncelleme, ajanın haber-tepki stratejisi üzerinden yapılır.

3.2. Haber Sistemi
Her adımda dışsal bir haber (pozitif/negatif/nötr) gelir.

Haberler, temel değeri etkileyen şoklar gibi modellenir:


Vt​ =Vt−1 +δt (haber etkisi)

Ajanlar, bu haberlerin gücüne göre pozisyon alır (alış/satış).

3.3. Piyasa Fiyatı
Fiyat, temel değere yakınsama eğilimindedir:

Pt=Pt−1 +α⋅(Vt − Pt−1)+ϵt​

𝛼: bilgi yansıtma katsayısı.

𝜖𝑡 : mikro gürültü/likidite etkisi.

| Senaryo                                                 | Açıklama                                                                                     |
| ------------------------------------------------------- | -------------------------------------------------------------------------------------------- |
| **Yalnızca temel ajanlar (homojen)**                    | Homojen duyarlılıklar: kolektif dengeye ulaşılır                                             |
| **Heterojen duyarlılık (α, risk, haber algısı farklı)** | Farklı fiyat tepkileri ve farklı kâr/zarar dağılımı gözlenir                                 |
| **Haber yoğunluğu arttıkça**                            | Piyasa daha sık dalgalanır ancak yine de temel değere yakınsar                               |
| **Yanıltıcı haber senaryosu**                           | Bazı ajanlar aşırı tepki verir, bazıları az etkilenir → bilgi kirliliği ile sapma oluşabilir |

```python
# Temel Ajanlar ve Haber Etkisi ile Evrimsel Piyasa Modeli


from mesa.datacollection import DataCollector
import random
import numpy as np
import matplotlib.pyplot as plt
################################################################################
"""
    Haberlerin rastgele olarak üretildiği ortamı simüle eder.
    news_prob: Haber gelme olasılığı.
    current_news: Mevcut haberin değeri (-1, 0, veya 1).
    step(): Her adımda yeni bir haber üretir.
"""

class NewsEnvironment:
    def __init__(self, news_prob=0.3):
        self.news_prob = news_prob  # haberin geldiği olasılık
        self.current_news = 0       # -1 (negatif), 0 (nötr), 1 (pozitif)

    def step(self):
        roll = random.random()
        if roll < self.news_prob:
            self.current_news = random.choice([-1, 1])
        else:
            self.current_news = 0
        return self.current_news

################################################################################
"""
    Piyasada işlem yapan temel ajanı temsil eder.

    risk_aversion: Ajanın riskten kaçınma seviyesi.

    sensitivity: Ajanın habere duyarlılığı.

    step(): Her adımda ajan haberlere göre alım veya satım kararı alır.

    get_wealth(): Ajanın toplam servetini hesaplar.
"""


class FundamentalAgent(mesa.Agent):
    def __init__(self,model, risk_aversion, sensitivity):
        super().__init__(model)
        self.cash = 1000
        self.shares = 0
        self.risk_aversion = risk_aversion  # [0, 1]
        self.sensitivity = sensitivity      # haber duyarlılığı [0, 1]

    def step(self):
        news = self.model.news.current_news
        expected_return = news * self.sensitivity

        # Karar mekanizması: risk duyarlılığına göre al/sat
        if expected_return > self.risk_aversion:
            # Alım yap
            price = self.model.price
            shares_to_buy = self.cash // price
            self.shares += shares_to_buy
            self.cash -= shares_to_buy * price
        elif expected_return < -self.risk_aversion:
            # Satış yap
            price = self.model.price
            self.cash += self.shares * price
            self.shares = 0

    def get_wealth(self):
        return self.cash + self.shares * self.model.price

################################################################################
"""
    Piyasa ortamını ve ajanların etkileşimini simüle eder.
    N: Ajan sayısı.
    price: Başlangıç fiyatı.
    news_prob: Haber gelme olasılığı.
    step(): Her adımda haberler üretilir, ajanlar karar alır ve fiyat güncellenir.
    update_price(): Piyasa fiyatını arz ve talebe göre günceller.
"""

class NewsMarketModel(mesa.Model):
    def __init__(self, N=50, price=100, news_prob=0.3):
        super().__init__()

        self.num_agents = N

        self.price = price
        self.news = NewsEnvironment(news_prob=news_prob)
        self.datacollector = DataCollector(
            model_reporters={"Price": lambda m: m.price},
            agent_reporters={"Wealth": lambda a: a.get_wealth()}
        )

        for i in range(self.num_agents):
            risk = random.uniform(0.2, 0.6)
            sens = random.uniform(0.3, 0.9)
            agent = FundamentalAgent(self, risk, sens)
            self.agents.add(agent)

    def step(self):
        self.news.step()
        self.agents.do("step")
        self.update_price()
        self.datacollector.collect(self)

    def update_price(self):
        net_demand = sum([a.shares for a in self.agents])
        self.price = max(1, 100 + np.tanh(net_demand / 10) * 10)
################################################################################
"""
    NewsMarketModel sınıfından bir model nesnesi oluşturulur.
    Simülasyon belirli sayıda adım için çalıştırılır.
    Toplanan veriler kullanılarak piyasa fiyatı ve ortalama servet grafikleri çizdirilir.

"""

# Simülasyonu çalıştıralım
model = NewsMarketModel(N=100, price=100, news_prob=0.4)
for i in range(100):
    model.step()

# Sonuçları gösterelim
data = model.datacollector.get_model_vars_dataframe()
agent_data = model.datacollector.get_agent_vars_dataframe()

plt.figure(figsize=(10, 4))
plt.plot(data["Price"])
plt.title("Piyasa Fiyatı (Sadece Temel Ajanlar ve Haber Etkisi)")
plt.xlabel("Adım")
plt.ylabel("Fiyat")
plt.grid(True)
plt.show()

# Ortalama servet
yeni_df = agent_data.reset_index()
average_wealth = yeni_df.groupby("Step")["Wealth"].mean()
plt.figure(figsize=(10, 4))
plt.plot(average_wealth)
plt.title("Ortalama Servet (Temel Ajanlar)")
plt.xlabel("Adım")
plt.ylabel("Servet")
plt.grid(True)
plt.show()
```

Modelimizi yukarıdaki Python kodu ile sadece temel ajanların ve haber etkisinin olduğu bir piyasa yapısı olarak tanımladık. Ajanların kararları, haberin pozitif ya da negatif oluşuna ve kendi haber duyarlılıklarına göre şekilleniyor. Bu yapı, teknik ya da gürültü etkilerinden arındırılmış bir temel değer odaklı piyasa davranışı sağlar.

Modele daha farklı haber yapıları (örneğin, pozitif-negatif duyarlılık ayrımı, kalıcı haber etkisi) ekleyebiliriz. Ve ardından sonuçları analiz edip yorumlayabiliriz

Pozitif/Negatif duyarlılık ayrımı ve kalıcı haber etkisi için modeli genişletelim.

Şimdi modelimizi pozitif-negatif haber duyarlılığı ayrımı ve kalıcı haber etkisi (decaying memory effect) ile genişletiyoruz. Bu yeni modelin detaylı tanımı yer aldığı  "sadece temel ajanlar varsa 1.tex" dosyaya bak.

Bu yapıyı kullanarak:

1. Simülasyonları çalıştırabiliriz (farklı $\theta^+ / \theta^-$ / $\gamma$ kombinasyonları),
2. Sonuçları grafiklerle analiz edebiliriz,
🧠 3. Duygu tabanlı yatırım davranışlarını yorumlayabiliriz.



```python
# Kodun yeniden çalıştırılması gerekiyor çünkü önceki kernel sıfırlandı
import numpy as np
import matplotlib.pyplot as plt

# Sabitler
np.random.seed(42)
T = 100  # Zaman adımı
N = 50   # Ajan sayısı

# Parametre aralıkları
theta_pos_values = [0.5, 1.0, 2.0]
theta_neg_values = [0.5, 1.0, 2.0]
gamma_values = [0.1, 0.5, 0.9]

# Haber sinyali (basit bir senaryo)
news_signal = np.random.normal(loc=0.0, scale=1.0, size=T)

# Simülasyon sonuçları için yapı
results = {}

for theta_pos in theta_pos_values:
    for theta_neg in theta_neg_values:
        for gamma in gamma_values:
            M = np.zeros((N, T))  # Bellek
            portfolio = np.zeros((N, T))  # Portföy değeri
            for t in range(1, T):
                M[:, t] = gamma * M[:, t - 1] + news_signal[t]
                positive_mask = M[:, t] >= 0
                delta_q = np.where(positive_mask,
                                   theta_pos * M[:, t],
                                   theta_neg * M[:, t])
                portfolio[:, t] = portfolio[:, t - 1] + delta_q
            # Ortalama portföy büyümesi kaydı
            key = f'θ⁺={theta_pos}, θ⁻={theta_neg}, γ={gamma}'
            results[key] = portfolio.mean(axis=0)

# Grafik çizimi
plt.figure(figsize=(14, 8))
for key, values in results.items():
    plt.plot(values, label=key)
plt.title("Ajan Portföy Performansı (Pozitif/Negatif Haber Duyarlılığı ve Bellek)")
plt.xlabel("Zaman")
plt.ylabel("Ortalama Portföy Değeri")
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True)
plt.tight_layout()
plt.show()
```

Grafikte, temel ajanların portföy büyüme performansları; pozitif haber duyarlılığı $\theta^+$, negatif haber duyarlılığı $\theta^-$ ve haber belleği katsayısı $\gamma$ parametrelerine göre analiz edilmiştir.

### Açıklamalar:

#### 1. **Pozitif ve Negatif Duyarlılıklar ($\theta^+, \theta^-$)**

* $\theta^+ > \theta^-$ olduğunda ajanlar pozitif haberlerden daha fazla etkilenir; portföy değerleri genellikle yükselir.
* $\theta^- > \theta^+$ olduğunda, kötü haberlerin etkisi baskın olur; ortalama büyüme düşer.

#### 2. **Bellek Katsayısı ($\gamma$)**

* Küçük $\gamma$ değerleri (örneğin 0.1), haber etkilerinin hızla unutulduğunu gösterir → tepkiler daha dalgalı ama kısa ömürlüdür.
* Büyük $\gamma$ değerleri (örneğin 0.9), haberlerin etkisinin kalıcı olduğunu gösterir → etkiler daha süreklidir, ancak birikimsel risk oluşturur.

Bir sonraki adım olarak bu modele daha gelişmiş haber yapıları (pozitif-negatif ayrımı, kalıcı haber etkisi, haber yoğunluğu, şok etkisi) ekleyebiliriz.

## 📘 3. Adım: Kalıcı Haber Etkisi ve Zamanla Azalan Haber Hafızası (LaTeX)

Bu adımda, haber etkisinin zamanla azaldığı ve ajanın bir süre boyunca geçmiş haberleri değerlendirdiği bir yapı kuruyoruz. Bu, **haberlerin kalıcılığı** ve **haber hafızası penceresi** kavramlarıyla modellenir.


### ✅ Sonraki Adım

| ---- | ---------------------------------------------------- |
| 1.   | Temel haber etkisi modelleme                         |

Şimdi bu modelleri kullanarak **simülasyonlar, grafik üretimi ve sonuçların karşılaştırılması** aşamasına geçebiliriz. Burada her yapının ayrı ayrı ve birlikte sistem üzerindeki etkisini kodla test edeceğiz. Başlayalım mı?





```python
import numpy as np
import matplotlib.pyplot as plt

# Simülasyon parametreleri
np.random.seed(42)
T = 100  # toplam adım
news = np.random.normal(0, 1, T)  # rastgele haber dizisi (ortalama 0, std 1)

# Haber etkisi modelleri
def basic_news_model(news):
    """Sadece mevcut habere göre karar."""
    return news.copy()

def asymmetric_news_model(news, alpha_pos=1.0, alpha_neg=1.5):
    """Pozitif ve negatif haber etkisine ayrı hassasiyet."""
    signal = []
    for n in news:
        if n >= 0:
            signal.append(alpha_pos * n)
        else:
            signal.append(alpha_neg * n)
    return np.array(signal)

def memory_news_model(news, lam=0.9, memory_length=10):
    """Zamana bağlı haber hafızası modeli."""
    signal = []
    for t in range(len(news)):
        memory = 0
        for k in range(1, memory_length+1):
            if t - k >= 0:
                memory += (lam ** k) * news[t - k]
        signal.append(memory)
    return np.array(signal)

# Tüm modellerin çıktısını üret
basic_signal = basic_news_model(news)
asym_signal = asymmetric_news_model(news)
memory_signal = memory_news_model(news)

# Grafik çizimi
plt.figure(figsize=(12, 6))
plt.plot(news, label="Orijinal Haber", linestyle='--', color='gray')
plt.plot(basic_signal, label="1. Temel Model")
plt.plot(asym_signal, label="2. Asimetrik Model")
plt.plot(memory_signal, label="3. Hafızalı Model")
plt.axhline(0, color='black', linewidth=0.5)
plt.title("Haber Tabanlı Ajan Tepkileri (Farklı Modeller)")
plt.xlabel("Zaman")
plt.ylabel("Haber Tabanlı Sinyal / Karar Etkisi")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

Yukarıdaki grafik, yalnızca temel ajanların bulunduğu bir piyasada üç farklı haber etkisi modelinin zaman içinde nasıl sinyal ürettiğini göstermektedir:

📊 Model Açıklamaları:
Temel Model (Mevcut haber):
Ajanlar sadece o andaki habere tepki verir. Pozitif haber pozitif sinyal, negatif haber negatif sinyal üretir. Hafıza veya duyarlılık yoktur.

Asimetrik Model:
Ajanlar negatif haberlere karşı daha duyarlıdır. Bu finansal literatürde sık görülen "kayıptan kaçınma" davranışıyla örtüşür. Negatif haberler daha güçlü tepki yaratır.

Hafızalı Model:
Ajanlar geçmiş haberleri de dikkate alır. Yakın geçmişteki haberler daha etkili, uzak geçmişteki haberler daha az etkilidir (üstel ağırlıkla). Bu, haberlerin kalıcılığı ve sürükleyici etkisini yakalar.

```python
import numpy as np
import matplotlib.pyplot as plt

# --- Parametreler --- #
np.random.seed(42)
T = 100  # toplam adım
N = 100  # ajan sayısı

# Haber üretimi (standart normal rastgele haberler)
news_series = np.random.normal(loc=0, scale=1, size=T)

# Ajan duyarlılık eşikleri
agent_thresholds = np.random.uniform(0.2, 1.0, size=N)

# Haber etkisi modelleri için sonuç depoları
price_basic = [100]
price_asymmetric = [100]
price_memory = [100]

# Asimetrik model parametreleri
alpha_pos = 1.0
alpha_neg = 1.5

# Hafızalı model parametreleri
lambda_decay = 0.9

# --- Simülasyon Başlangıcı --- #
for t in range(1, T):
    H_t = news_series[t]

    # 1. Temel Model
    actions_basic = []
    for i in range(N):
        if H_t > agent_thresholds[i]:
            actions_basic.append(1)
        elif H_t < -agent_thresholds[i]:
            actions_basic.append(-1)
        else:
            actions_basic.append(0)
    price_change_basic = sum(actions_basic) / N
    price_basic.append(price_basic[-1] + price_change_basic)

    # 2. Asimetrik Model
    actions_asym = []
    for i in range(N):
        if H_t > 0 and alpha_pos * H_t > agent_thresholds[i]:
            actions_asym.append(1)
        elif H_t < 0 and alpha_neg * H_t < -agent_thresholds[i]:
            actions_asym.append(-1)
        else:
            actions_asym.append(0)
    price_change_asym = sum(actions_asym) / N
    price_asymmetric.append(price_asymmetric[-1] + price_change_asym)

    # 3. Hafızalı Model
    memory_window = news_series[max(0, t - 10):t + 1][::-1]  # son 10 haber
    weights = np.array([lambda_decay ** i for i in range(len(memory_window))])
    H_mem = np.dot(memory_window, weights) / sum(weights)
    actions_mem = []
    for i in range(N):
        if H_mem > agent_thresholds[i]:
            actions_mem.append(1)
        elif H_mem < -agent_thresholds[i]:
            actions_mem.append(-1)
        else:
            actions_mem.append(0)
    price_change_mem = sum(actions_mem) / N
    price_memory.append(price_memory[-1] + price_change_mem)

# --- Grafikler --- #
plt.figure(figsize=(12, 6))
plt.plot(price_basic, label="Temel Model")
plt.plot(price_asymmetric, label="Asimetrik Model")
plt.plot(price_memory, label="Hafızalı Model")
plt.title("Fiyat Dinamikleri - Farklı Haber Etkileri")
plt.xlabel("Zaman")
plt.ylabel("Fiyat")
plt.legend()
plt.grid(True)
plt.show()
```

Modelin üç versiyonunu tek bir yapı altında tanımladık:

Temel Model: Haberler pozitifse alım, negatifse satım.

Asimetrik Model: Negatif haberlere karşı daha güçlü tepki.

Hafızalı Model: Son haberlerin ağırlıklı ortalaması ile karar.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Simülasyon parametreleri
np.random.seed(42)
steps = 200
news = np.random.choice([-1, 1], size=steps, p=[0.5, 0.5])  # Rastgele haberler

# Model 1: Temel model
def model_1(news):
    price = [100]
    for n in news:
        delta = 1 if n > 0 else -1
        price.append(price[-1] + delta)
    return price

# Model 2: Asimetrik model (negatif haber 2x etki)
def model_2(news):
    price = [100]
    for n in news:
        if n > 0:
            delta = 1
        else:
            delta = -2
        price.append(price[-1] + delta)
    return price

# Model 3: Hafızalı model (son 5 haberin ağırlıklı ortalamasına göre)
def model_3(news, window=5):
    price = [100]
    for t in range(len(news)):
        if t < window:
            avg_signal = np.mean(news[:t+1])
        else:
            weights = np.arange(1, window + 1)
            avg_signal = np.average(news[t-window+1:t+1], weights=weights)
        delta = np.sign(avg_signal)
        price.append(price[-1] + delta)
    return price

# Fiyat dizilerini hesapla
p1 = model_1(news)
p2 = model_2(news)
p3 = model_3(news)

# Sonuçları birleştir
df = pd.DataFrame({
    "Step": np.arange(steps+1),
    "Model 1 (Temel)": p1,
    "Model 2 (Asimetrik)": p2,
    "Model 3 (Hafızalı)": p3
})

# Grafik oluştur
plt.figure(figsize=(12, 6))
sns.lineplot(data=df.set_index("Step"))
plt.title("Temel, Asimetrik ve Hafızalı Modellerin Fiyat Simülasyonu")
plt.ylabel("Fiyat")
plt.xlabel("Zaman Adımı")
plt.grid(True)
plt.tight_layout()
plt.show()

# Temel istatistikler
summary_stats = df.drop(columns="Step").describe().T
summary_stats["Volatilite"] = df.drop(columns="Step").diff().std().values
summary_stats
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Simülasyon parametreleri
T = 100  # zaman adımı
np.random.seed(42)
news_stream = np.random.choice([-1, 0, 1], size=T, p=[0.3, 0.4, 0.3])

# Ortak başlangıç fiyatı
P0 = 100

# Model 1: Sabit pozitif-negatif etkili temel model
def model_1(news, alpha=1):
    P = [P0]
    for s in news:
        P.append(P[-1] + alpha * s)
    return P

# Model 2: Sabit haber etkili temel model (sadece yön etkisi)
def model_2(news):
    P = [P0]
    for s in news:
        if s != 0:
            P.append(P[-1] + 1 * np.sign(s))
        else:
            P.append(P[-1])
    return P

# Model 3: Duyarlılıkla güçlenen etki
def model_3(news):
    P = [P0]
    score = 0
    for s in news:
        score = score * 0.9 + s
        P.append(P[-1] + score)
    return P

# Model 4: Asimetrik duyarlılık
def model_4(news, alpha=1, beta=2):
    P = [P0]
    for s in news:
        if s > 0:
            P.append(P[-1] + alpha)
        elif s < 0:
            P.append(P[-1] - beta)
        else:
            P.append(P[-1])
    return P

# Model 5: Sönümlü haber etkisi
def model_5(news, lam=0.8, N=5):
    P = [P0]
    for t in range(len(news)):
        effect = 0
        for i in range(N):
            if t - i >= 0:
                effect += (lam ** i) * news[t - i]
        P.append(P[-1] + effect)
    return P

# Tüm modelleri çalıştır
price_1 = model_1(news_stream)
price_2 = model_2(news_stream)
price_3 = model_3(news_stream)
price_4 = model_4(news_stream)
price_5 = model_5(news_stream)

# Grafik çizimi
plt.figure(figsize=(12, 7))
plt.plot(price_1, label="Model 1: Doğrusal")
plt.plot(price_2, label="Model 2: Yön Temelli")
plt.plot(price_3, label="Model 3: Güçlenen Etki")
plt.plot(price_4, label="Model 4: Asimetrik")
plt.plot(price_5, label="Model 5: Sönümlü Etki")
plt.title("Farklı Haber Duyarlılığı Modelleri ile Fiyat Dinamikleri")
plt.xlabel("Zaman")
plt.ylabel("Fiyat")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

Yukarıdaki grafik, farklı haber duyarlılığı modellerine göre fiyatların zaman içindeki evrimini göstermektedir. Her model, haberlerin piyasaya etkisini farklı bir yaklaşımla temsil eder:

Kısa Açıklamalar:
Model 1 (Doğrusal): Pozitif ve negatif haberler simetrik ve sabit bir şekilde fiyatı etkiler.

Model 2 (Yön Temelli): Yalnızca haberin yönü önemlidir, etkisi sabittir.

Model 3 (Güçlenen Etki): Duyarlılık zamanla birikir ve artan ivmeyle fiyatı etkiler.

Model 4 (Asimetrik): Negatif haberler pozitiflere kıyasla daha büyük etki yaratır.

Model 5 (Sönümlü Etki): Geçmiş haberler giderek azalan ağırlıkla etkili olur (hafıza etkisi).
